````python
import requests
import re
import json

def extract_json(text):
    # remove markdown ```json ``` blocks
    text = re.sub(r"```json|```", "", text)
    
    # extract JSON array
    match = re.search(r"\[.*\]", text, re.DOTALL)
    if match:
        return match.group(0)
    
    return text


def prompt_llm(prompt, model="llama3.1:8b", temperature=0.1):
    res = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": model,
            "prompt": prompt,
            "stream": False,
            "options": {
                "temperature": temperature
            }
        }
    )
    return res.json()["response"]

# def chat_llm(messages, model="llama3.1:8b"):
#     res = requests.post(
#         "http://localhost:11434/api/chat",
#         json={
#             "model": model,
#             "messages": messages,
#             "stream": False,
#         }
#     )
#     return res.json()["message"]["content"]



def chat_llm(messages, temperature=0.1, model="llama3.1:8b"):
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": model,
            "messages": messages,
            "stream": False,
            "options": {
                "temperature": temperature
            }
        }
    )
    return res.json()["message"]["content"]



def chat_stream(messages, model="llama3.1:8b"):
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": model,
            "messages": messages,
            "stream": True
        },
        stream=True
    )

    full_response = ""

    for line in res.iter_lines():
        if line:
            data = json.loads(line.decode("utf-8"))
            chunk = data.get("message", {}).get("content", "")
            
            print(chunk, end="", flush=True)  # 👈 streaming effect
            full_response += chunk

    print()  # newline
    return full_response
````

````python
def ask_structured(messages, max_retries=3):
    for i in range(max_retries):
        response = chat_llm(messages)
        cleaned = extract_json(response)

        try:
            data = json.loads(cleaned)
            return data
        except Exception as e:
            print(f"Retry {i+1} بسبب:", e)

            # 🔥 reinforce instruction
            messages.append({
                "role": "user",
                "content": "Your previous output was invalid JSON. Fix it."
            })

    raise ValueError("Failed after retries")
````

````python
import datetime
import uuid

def get_current_time():
    return datetime.datetime.now().strftime("%H:%M:%S")

def fake_llm_generate_calls(user_input):
    user_input = user_input.lower()

    tool_calls = []

    # Case 1: math
    if any(op in user_input for op in ["+", "-", "*", "/"]):
        expressions = ["10+10", "30+30"]  # simplified parsing

        for expr in expressions:
            tool_calls.append({
                "tool": "calculator",
                "tool_id": str(uuid.uuid4()),
                "arguments": {"expression": expr}
            })

    # Case 2: time
    elif "time" in user_input:
        tool_calls.append({
            "tool": "get_time",
            "tool_id": str(uuid.uuid4()),
            "arguments": {}
        })

    # Case 3: no tool needed
    else:
        return None  # means LLM can answer directly

    return tool_calls

def execute_tool_calls(tool_calls):
    results = []

    for call in tool_calls:
        tool_id = call["tool_id"]
        tool = call["tool"]

        if tool == "calculator":
            output = eval(call["arguments"]["expression"])

        elif tool == "get_time":
            output = get_current_time()

        else:
            output = "Unknown tool"

        results.append({
            "tool_id": tool_id,
            "content": output
        })

    return results

def fake_llm_direct_answer(user_input):
    return "I can answer this without tools."

def fake_llm_final_answer(tool_calls, tool_results):
    # Map tool_id → result
    result_map = {r["tool_id"]: r["content"] for r in tool_results}

    answers = []

    for call in tool_calls:
        expr = call["arguments"].get("expression", "unknown")
        tool_id = call["tool_id"]
        result = result_map.get(tool_id, "N/A")

        # Handle different tools
        if call["tool"] == "calculator":
            answers.append(f"{expr} = {result}")
        elif call["tool"] == "get_time":
            answers.append(f"Current time = {result}")

    return "\n".join(answers)


def run_agent(user_input):
    print("User:", user_input)

    tool_calls = fake_llm_generate_calls(user_input)

    # CASE 1: No tool needed
    if tool_calls is None:
        answer = fake_llm_direct_answer(user_input)
        print("\nFinal Answer:")
        print(answer)
        return

    print("\nLLM Tool Calls:")
    print(tool_calls)

    tool_results = execute_tool_calls(tool_calls)

    print("\nTool Results:")
    print(tool_results)

    final_answer = fake_llm_final_answer(tool_calls, tool_results)

    print("\nFinal Answer:")
    print(final_answer)
````

````python
"""
simple_tool_demo.py
-------------------
Understand how tool calling works step by step.
"""

import json
import requests

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "llama3.1:8b"


def chat(messages, tools=None):
    payload = {"model": MODEL, "messages": messages, "stream": False}
    if tools:
        payload["tools"] = tools
    res = requests.post(OLLAMA_URL, json=payload)
    return res.json()["message"]


# =============================================================================
# Define a simple calculator tool
# =============================================================================

calculator_tool = {
    "type": "function",
    "function": {
        "name": "calculator",
        "description": "Use this to evaluate math expressions.",
        "parameters": {
            "type": "object",
            "properties": {
                "expression": {"type": "string", "description": "Math expression e.g. 10+10"}
            },
            "required": ["expression"]
        }
    }
}


# =============================================================================
# TEST 1: No tools passed
# Model answers from its own knowledge — no tool call possible
# =============================================================================
print("=" * 50)
print("TEST 1: No tools passed")
print("=" * 50)

response = chat([{"role": "user", "content": "What is 10+10?"}])
print("tool_calls:", response.get("tool_calls"))   # None
print("content:   ", response.get("content"))      # "20"


# =============================================================================
# TEST 2: Tool passed, model decides freely
# Model may or may not call the tool — it's its own choice
# =============================================================================
print("\n" + "=" * 50)
print("TEST 2: Tool passed, model decides freely")
print("=" * 50)

response = chat(
    messages=[{"role": "user", "content": "What is 10+10?"}],
    tools=[calculator_tool]
)
print("tool_calls:", response.get("tool_calls"))   # may be None or a call
print("content:   ", response.get("content"))


# =============================================================================
# TEST 3: Force tool call via system prompt
# Model MUST call the tool — we enforce it
# =============================================================================
print("\n" + "=" * 50)
print("TEST 3: Force tool call via system prompt")
print("=" * 50)

response = chat(
    messages=[
        {"role": "system", "content": "You MUST call the 'calculator' tool. Never answer in plain text."},
        {"role": "user",   "content": "What is 10+10?"}
    ],
    tools=[calculator_tool]
)
print("tool_calls:", response.get("tool_calls"))   # should be a call
print("content:   ", response.get("content"))

# If tool was called, show the args
tool_calls = response.get("tool_calls", [])
if tool_calls:
    args = tool_calls[0]["function"]["arguments"]
    if isinstance(args, str):
        args = json.loads(args)
    print("args:      ", args)
    # Now execute it
    result = eval(args["expression"])
    print("result:    ", result)
````

## Implementing multiple tools

````python
import requests
import json
import datetime
import uuid

# =========================
# 1. TOOL REGISTRY
# =========================
TOOLS = [
    {
        "name": "calculator",
        "description": "Perform mathematical calculations",
        "parameters": {
            "expression": "string"
        }
    },
    {
        "name": "get_time",
        "description": "Get current time",
        "parameters": {}
    }
]


# =========================
# 2. TOOL IMPLEMENTATIONS
# =========================
def calculator(expression):
    try:
        return eval(expression)
    except Exception as e:
        raise Exception(str(e))


def get_time():
    return datetime.datetime.now().strftime("%H:%M:%S")


def run_tool(name, args):
    if name == "calculator":
        return calculator(args.get("expression", ""))

    elif name == "get_time":
        return get_time()

    else:
        raise Exception(f"Unknown tool: {name}")


# =========================
# 3. TOOL EXECUTION
# =========================
def run_tools(tool_calls):
    results = []

    for call in tool_calls:
        tool_id = call.get("tool_id")
        tool_name = call.get("tool")
        args = call.get("arguments", {})

        try:
            output = run_tool(tool_name, args)

            results.append({
                "type": "tool_result",
                "tool_use_id": tool_id,
                "content": json.dumps(output),
                "is_error": False
            })

        except Exception as e:
            results.append({
                "type": "tool_result",
                "tool_use_id": tool_id,
                "content": str(e),
                "is_error": True
            })

    return results


# =========================
# 4. TOOL PROMPT BUILDER
# =========================
def build_tool_prompt(tools):
    text = ""

    for t in tools:
        text += f"""
Tool: {t['name']}
Description: {t['description']}
Parameters: {json.dumps(t['parameters'])}
"""

    return text


# =========================
# 5. SIMPLE ROUTER (OPTIONAL)
# =========================
def select_tools(user_input):
    user_input = user_input.lower()
    selected = []

    if any(op in user_input for op in ["+", "-", "*", "/"]):
        selected.append("calculator")

    if "time" in user_input:
        selected.append("get_time")

    # fallback → give all tools
    if not selected:
        return TOOLS

    return [t for t in TOOLS if t["name"] in selected]


# =========================
# 6. LLM CALL
# =========================
def chat_llm(messages, model="llama3.1:8b"):
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": model,
            "messages": messages,
            "stream": False
        }
    )
    return res.json()["message"]["content"]


# =========================
# 7. JSON PARSER
# =========================
def try_parse_json(text):
    try:
        return json.loads(text)
    except:
        return None


# =========================
# 8. MAIN AGENT LOOP
# =========================
def run_conversation(user_input):

    # 🔥 dynamic tool selection
    selected_tools = select_tools(user_input)

    system_prompt = f"""
                    You are an AI assistant that can use tools.

                    Available tools:
                    {build_tool_prompt(selected_tools)}

                    RULES:

                    STEP 1:
                    - If tools are needed → return ONLY JSON:
                    {{
                      "tool_calls": [
                        {{
                          "tool": "tool_name",
                          "tool_id": "unique_id",
                          "arguments": {{...}}
                        }}
                      ]
                    }}

                    STEP 2:
                    - After receiving tool results → return FINAL ANSWER (no JSON)

                    GENERAL:
                    - Do NOT mix JSON and text
                    - Do NOT explain tool usage
                    """

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_input}
    ]

    step = 0

    while True:
        step += 1
        print(f"\n--- STEP {step} ---")

        response = chat_llm(messages)
        print("\nLLM:", response)

        data = try_parse_json(response)

        # =========================
        # TOOL CALL CASE
        # =========================
        if isinstance(data, dict) and "tool_calls" in data:

            tool_calls = data["tool_calls"]

            # enforce tool_id
            for i, call in enumerate(tool_calls):
                if "tool_id" not in call:
                    call["tool_id"] = f"call_{i+1}"

            print("\nTool Calls:")
            print(tool_calls)

            tool_results = run_tools(tool_calls)

            print("\nTool Results:")
            print(tool_results)

            # add assistant response
            messages.append({
                "role": "assistant",
                "content": response
            })

            # feed results back
            messages.append({
                "role": "user",
                "content": f"""
                            Tool results:
                            {json.dumps(tool_results)}

                            Now use these results to answer the question.
                            """
            })

        # =========================
        # FINAL ANSWER
        # =========================
        else:
            print("\nFinal Answer:")
            print(response)
            break
````

## Using multiple tools

````python
import requests
import json
import datetime

# =========================
# 1. TOOL IMPLEMENTATIONS
# =========================

def calculator(expression: str) -> str:
    try:
        return str(eval(expression, {"__builtins__": {}}))
    except Exception as e:
        return f"ERROR: {e}"

def get_current_time() -> str:
    return datetime.datetime.now().strftime("%H:%M:%S")

def add_hours(time_str: str, hours: float) -> str:
    dt = datetime.datetime.strptime(time_str, "%H:%M:%S")
    result = dt + datetime.timedelta(hours=float(hours))
    return result.strftime("%H:%M:%S")

def add_duration_to_datetime(date_str: str, days: int) -> str:
    dt = datetime.datetime.strptime(date_str, "%Y-%m-%d")
    result = dt + datetime.timedelta(days=int(days))
    return result.strftime("%Y-%m-%d")

def set_reminder(date_str: str, content: str) -> str:
    msg = f"[REMINDER SET] Date: {date_str} | Content: {content}"
    print(msg)
    return msg

TOOL_REGISTRY = {
    "calculator":              calculator,
    "get_current_time":        get_current_time,
    "add_hours":               add_hours,
    "add_duration_to_datetime": add_duration_to_datetime,
    "set_reminder":            set_reminder,
}

def run_tool(name: str, args: dict) -> str:
    if name not in TOOL_REGISTRY:
        raise ValueError(f"Unknown tool: {name}")
    return str(TOOL_REGISTRY[name](**args))


# =========================
# 2. LLM CALL (plain chat, no tool API)
# =========================

def chat_llm(messages: list, model: str = "llama3.1:8b") -> str:
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={"model": model, "messages": messages, "stream": False},
    )
    res.raise_for_status()
    return res.json()["message"]["content"]


# =========================
# 3. PLANNER PROMPT
#    Ask the LLM to produce a JSON plan — nothing else.
# =========================

PLANNER_SYSTEM = """You are a planner that converts user requests into tool execution plans.

Available tools:
- calculator        → args: {"expression": "<math expression as string>"}
- get_current_time  → args: {}
- add_hours         → args: {"time_str": "<HH:MM:SS>", "hours": <number>}
- add_duration_to_datetime → args: {"date_str": "<YYYY-MM-DD>", "days": <integer>}
- set_reminder      → args: {"date_str": "<YYYY-MM-DD>", "content": "<string>"}

Rules:
1. Return ONLY a valid JSON array. No explanation, no markdown, no backticks.
2. Each step: {"step": <int>, "tool": "<name>", "args": {<key>: <value>}}
3. To use a previous step's output as an argument, write "$step_N" as the value (e.g. "$step_1").
4. NEVER pass a "$step_N" reference into calculator — calculator only accepts plain math strings.
5. NEVER add extra steps. Use the minimum steps needed. Do not add intermediate calculator steps when a dedicated tool exists.
6. add_hours handles time arithmetic — do NOT use calculator for time.
7. set_reminder content must never be empty — infer a short description from context.

Examples:

User: What is 10+10?
[{"step":1,"tool":"calculator","args":{"expression":"10+10"}}]

User: What time is it now and what will it be 2 hours from now?
[{"step":1,"tool":"get_current_time","args":{}},{"step":2,"tool":"add_hours","args":{"time_str":"$step_1","hours":2}}]

User: What is the current time?
[{"step":1,"tool":"get_current_time","args":{}}]

User: Set a reminder 177 days after 2050-01-01
[{"step":1,"tool":"add_duration_to_datetime","args":{"date_str":"2050-01-01","days":177}},{"step":2,"tool":"set_reminder","args":{"date_str":"$step_1","content":"Scheduled reminder"}}]

User: What is 15 * 4 and then add 10?
[{"step":1,"tool":"calculator","args":{"expression":"15*4+10"}}]
"""

def make_plan(user_input: str, model: str) -> list:
    messages = [
        {"role": "system",  "content": PLANNER_SYSTEM},
        {"role": "user",    "content": user_input},
    ]
    raw = chat_llm(messages, model=model)

    # Strip any accidental markdown fences
    clean = raw.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()

    try:
        plan = json.loads(clean)
        assert isinstance(plan, list)
        return validate_plan(plan)
    except ValueError:
        raise
    except Exception:
        raise ValueError(f"Planner returned invalid JSON:\n{raw}")


def validate_plan(plan: list) -> list:
    """
    Remove hallucinated steps that would cause runtime errors.
    Specifically: drop any calculator step whose expression contains a $step_N reference
    (the LLM sometimes inserts a bogus intermediate calculator step for time arithmetic).
    """
    valid_tools = set(TOOL_REGISTRY.keys())
    cleaned = []
    for step in plan:
        tool = step.get("tool", "")
        args = step.get("args", {})

        # Drop unknown tools
        if tool not in valid_tools:
            print(f"  [validator] Dropping unknown tool step: {tool}")
            continue

        # Drop calculator steps where expression contains a $step reference
        if tool == "calculator":
            expr = args.get("expression", "")
            if "$step_" in str(expr):
                print(f"  [validator] Dropping invalid calculator step: {expr!r}")
                continue

        cleaned.append(step)

    # Re-number steps sequentially after filtering
    for i, step in enumerate(cleaned, start=1):
        step["step"] = i

    return cleaned


# =========================
# 4. EXECUTOR
#    Resolves $step_N references and runs each step in order.
# =========================

def execute_plan(plan: list) -> dict:
    results = {}
    for step in plan:
        step_id = int(step["step"])
        tool    = step["tool"]
        args    = dict(step.get("args", {}))  # copy so we can mutate

        # Resolve $step_N references using whatever step produced that result
        for key, val in args.items():
            if isinstance(val, str) and val.startswith("$step_"):
                ref = int(val.replace("$step_", ""))
                if ref not in results:
                    raise ValueError(f"Step {step_id} references $step_{ref} but that step hasn't run yet or was removed.")
                args[key] = results[ref]

        print(f"  Step {step_id} → {tool}({args})")
        output = run_tool(tool, args)
        print(f"           ← {output}")
        results[step_id] = output

    return results


# =========================
# 5. SUMMARISER
#    Ask the LLM to turn raw results into a human-friendly answer.
# =========================

SUMMARISER_SYSTEM = "You are a helpful assistant. Given a user request and the results of tool calls, write a clear, concise answer in plain English. No JSON, no code."

def summarise(user_input: str, plan: list, results: dict, model: str) -> str:
    steps_text = "\n".join(
        f"Step {s['step']} ({s['tool']}) → {results.get(s['step'], '?')}"
        for s in plan
    )
    messages = [
        {"role": "system",  "content": SUMMARISER_SYSTEM},
        {"role": "user",    "content": f"User asked: {user_input}\n\nTool results:\n{steps_text}"},
    ]
    return chat_llm(messages, model=model)


# =========================
# 6. MAIN ENTRY POINT
# =========================

def run_conversation(user_input: str, model: str = "llama3.1:8b"):
    print(f"\n{'='*50}")
    print(f"USER: {user_input}")
    print(f"{'='*50}")

    # Step A: Plan
    try:
        plan = make_plan(user_input, model)
    except ValueError as e:
        print(f"❌ Planning failed: {e}")
        return None
    print(f"\nPLAN: {json.dumps(plan, indent=2)}")

    # Step B: Execute
    print("\nEXECUTION:")
    try:
        results = execute_plan(plan)
    except Exception as e:
        print(f"❌ Execution failed: {e}")
        return None

    # Step C: Summarise
    final = summarise(user_input, plan, results, model)
    print(f"\nFINAL ANSWER: {final}")
    return final


if __name__ == "__main__":
    run_conversation("What is 10+10?")
    run_conversation("What time is it now and what will it be 2 hours from now?")
    run_conversation("Please set a reminder that is 177 days after January 1st, 2050.")
````

````python
"""
ollama_agent.py
---------------
A data-driven agentic loop for Ollama/Llama.
Adding a new tool = one entry in TOOLS. Nothing else changes.
"""

import requests
import json
import datetime
import inspect
from typing import Callable


# =============================================================================
# 1. TOOL DEFINITIONS
#    Each tool is a plain dict with:
#      fn          – the callable
#      description – one sentence for the planner
#      args        – {param_name: "type hint + description"} for prompt generation
#      no_ref      – args that must NOT receive a $step_N reference (e.g. calculator expression)
# =============================================================================

def _calculator(expression: str) -> str:
    try:
        return str(eval(expression, {"__builtins__": {}}))
    except Exception as e:
        return f"ERROR: {e}"

def _get_current_time() -> str:
    return datetime.datetime.now().strftime("%H:%M:%S")

def _add_hours(time_str: str, hours: float) -> str:
    dt = datetime.datetime.strptime(time_str, "%H:%M:%S")
    return (dt + datetime.timedelta(hours=float(hours))).strftime("%H:%M:%S")

def _add_duration_to_datetime(date_str: str, days: int) -> str:
    dt = datetime.datetime.strptime(date_str, "%Y-%m-%d")
    return (dt + datetime.timedelta(days=int(days))).strftime("%Y-%m-%d")

def _set_reminder(date_str: str, content: str) -> str:
    msg = f"[REMINDER SET] Date: {date_str} | Content: {content}"
    print(msg)
    return msg


# ── Single source of truth ───────────────────────────────────────────────────
TOOLS: list[dict] = [
    {
        "name":        "calculator",
        "fn":          _calculator,
        "description": "Evaluate a plain math expression and return the result.",
        "args":        {"expression": 'str — math expression e.g. "10+10" or "177*2"'},
        "no_ref":      {"expression"},          # $step_N must never land here
    },
    {
        "name":        "get_current_time",
        "fn":          _get_current_time,
        "description": "Return the current local time as HH:MM:SS.",
        "args":        {},
    },
    {
        "name":        "add_hours",
        "fn":          _add_hours,
        "description": "Add N hours to a HH:MM:SS time string and return the new time.",
        "args":        {
            "time_str": "str — starting time in HH:MM:SS format",
            "hours":    "float — hours to add",
        },
    },
    {
        "name":        "add_duration_to_datetime",
        "fn":          _add_duration_to_datetime,
        "description": "Add N days to a YYYY-MM-DD date string and return the new date.",
        "args":        {
            "date_str": "str — starting date in YYYY-MM-DD format",
            "days":     "int — days to add",
        },
    },
    {
        "name":        "set_reminder",
        "fn":          _set_reminder,
        "description": "Set a reminder on a given date. Content must never be empty.",
        "args":        {
            "date_str": "str — reminder date in YYYY-MM-DD format",
            "content":  "str — reminder message, infer from context if not given",
        },
    },
]

# Derived lookups — built automatically, never written by hand
_TOOL_BY_NAME: dict[str, dict] = {t["name"]: t for t in TOOLS}


# =============================================================================
# 2. EXECUTION
# =============================================================================

def run_tool(name: str, args: dict) -> str:
    tool = _TOOL_BY_NAME.get(name)
    if tool is None:
        raise ValueError(f"Unknown tool: {name!r}. Available: {list(_TOOL_BY_NAME)}")
    return str(tool["fn"](**args))


def execute_plan(plan: list) -> dict:
    results = {}
    for step in plan:
        step_id = int(step["step"])
        tool    = step["tool"]
        args    = dict(step.get("args", {}))

        for key, val in args.items():
            if isinstance(val, str) and val.startswith("$step_"):
                ref = int(val.replace("$step_", ""))
                if ref not in results:
                    raise ValueError(
                        f"Step {step_id} references $step_{ref} "
                        f"which hasn't run yet or was removed."
                    )
                args[key] = results[ref]

        print(f"  Step {step_id} → {tool}({args})")
        output = run_tool(tool, args)
        print(f"           ← {output}")
        results[step_id] = output

    return results


# =============================================================================
# 3. PLANNER  —  prompt is generated from TOOLS, not written by hand
# =============================================================================

def _build_tool_docs() -> str:
    """Generate the tool-listing section of the planner prompt from TOOLS."""
    lines = []
    for t in TOOLS:
        arg_str = json.dumps({k: f"<{v}>" for k, v in t["args"].items()}) if t["args"] else "{}"
        lines.append(f'- {t["name"]:30s} — {t["description"]}')
        lines.append(f'  args: {arg_str}')
        if t.get("no_ref"):
            lines.append(f'  ⚠ These args must NOT be "$step_N" references: {sorted(t["no_ref"])}')
    return "\n".join(lines)


def _build_planner_system() -> str:
    return f"""You are a planner. Convert user requests into a minimal JSON execution plan.

TOOLS
-----
{_build_tool_docs()}

RULES
-----
1. Return ONLY a valid JSON array — no explanation, no markdown, no backticks.
2. Each element: {{"step": <int>, "tool": "<name>", "args": {{...}}}}
3. To forward one step's output to a later step, use "$step_N" as the arg value.
4. Never pass "$step_N" into an arg marked with ⚠ above.
5. Use the minimum number of steps. Prefer dedicated tools over calculator for non-math tasks.
6. set_reminder content must never be empty — infer from context.

EXAMPLES
--------
User: What is 10+10?
[{{"step":1,"tool":"calculator","args":{{"expression":"10+10"}}}}]

User: What time is it now and what will it be 2 hours from now?
[{{"step":1,"tool":"get_current_time","args":{{}}}},{{"step":2,"tool":"add_hours","args":{{"time_str":"$step_1","hours":2}}}}]

User: Set a reminder 177 days after 2050-01-01.
[{{"step":1,"tool":"add_duration_to_datetime","args":{{"date_str":"2050-01-01","days":177}}}},{{"step":2,"tool":"set_reminder","args":{{"date_str":"$step_1","content":"Scheduled reminder"}}}}]
"""


def _validate_plan(plan: list) -> list:
    """Drop steps with unknown tools or illegal $step_N references, then renumber."""
    cleaned = []
    for step in plan:
        tool_name = step.get("tool", "")
        tool_def  = _TOOL_BY_NAME.get(tool_name)
        args      = step.get("args", {})

        if tool_def is None:
            print(f"  [validator] Dropping unknown tool: {tool_name!r}")
            continue

        bad_refs = tool_def.get("no_ref", set())
        skip = False
        for arg_name in bad_refs:
            val = args.get(arg_name, "")
            if isinstance(val, str) and "$step_" in val:
                print(f"  [validator] Dropping step with illegal $step ref in {arg_name!r}: {val!r}")
                skip = True
                break
        if skip:
            continue

        cleaned.append(step)

    for i, step in enumerate(cleaned, start=1):
        step["step"] = i

    return cleaned


def make_plan(user_input: str, model: str) -> list:
    system = _build_planner_system()
    messages = [
        {"role": "system", "content": system},
        {"role": "user",   "content": user_input},
    ]
    raw   = chat_llm(messages, model=model)
    clean = raw.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()
    try:
        plan = json.loads(clean)
        assert isinstance(plan, list)
        return _validate_plan(plan)
    except (json.JSONDecodeError, AssertionError):
        raise ValueError(f"Planner returned invalid JSON:\n{raw}")


# =============================================================================
# 4. LLM CALL
# =============================================================================

def chat_llm(messages: list, model: str = "llama3.1:8b") -> str:
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={"model": model, "messages": messages, "stream": False},
    )
    res.raise_for_status()
    return res.json()["message"]["content"]


# =============================================================================
# 5. SUMMARISER
# =============================================================================

_SUMMARISER_SYSTEM = (
    "You are a helpful assistant. Given a user request and tool results, "
    "write a clear, concise answer in plain English. No JSON, no code."
)

def summarise(user_input: str, plan: list, results: dict, model: str) -> str:
    steps_text = "\n".join(
        f"  Step {s['step']} ({s['tool']}) → {results.get(s['step'], '?')}"
        for s in plan
    )
    messages = [
        {"role": "system", "content": _SUMMARISER_SYSTEM},
        {"role": "user",   "content": f"User asked: {user_input}\n\nTool results:\n{steps_text}"},
    ]
    return chat_llm(messages, model=model)


# =============================================================================
# 6. MAIN
# =============================================================================

def run_conversation(user_input: str, model: str = "llama3.1:8b"):
    print(f"\n{'='*50}")
    print(f"USER: {user_input}")
    print(f"{'='*50}")

    try:
        plan = make_plan(user_input, model)
    except ValueError as e:
        print(f"❌ Planning failed: {e}")
        return None
    print(f"\nPLAN: {json.dumps(plan, indent=2)}")

    print("\nEXECUTION:")
    try:
        results = execute_plan(plan)
    except Exception as e:
        print(f"❌ Execution failed: {e}")
        return None

    final = summarise(user_input, plan, results, model)
    print(f"\nFINAL ANSWER: {final}")
    return final


if __name__ == "__main__":
    run_conversation("What is 10+10?")
    run_conversation("What time is it now and what will it be 2 hours from now?")
    run_conversation("Please set a reminder that is 177 days after January 1st, 2050.")
````

## The batch tool

````python
"""
ollama_agent.py
---------------
A data-driven agentic loop for Ollama/Llama.
Adding a new tool = one entry in TOOLS. Nothing else changes.

Batch tool trick (from transcript):
  Small LLMs rarely emit multiple parallel steps on their own.
  The batch tool is a meta-tool that tells the planner "group these
  independent calls together". The executor then fans them out itself.
  This is the Ollama equivalent of Claude's multi-tool-use block pattern.
"""

import json
import datetime
import requests


# =============================================================================
# 1. TOOL IMPLEMENTATIONS
# =============================================================================

def _calculator(expression: str) -> str:
    try:
        return str(eval(expression, {"__builtins__": {}}))
    except Exception as e:
        return f"ERROR: {e}"

def _get_current_time() -> str:
    return datetime.datetime.now().strftime("%H:%M:%S")

def _add_hours(time_str: str, hours: float) -> str:
    dt = datetime.datetime.strptime(time_str, "%H:%M:%S")
    return (dt + datetime.timedelta(hours=float(hours))).strftime("%H:%M:%S")

def _add_duration_to_datetime(date_str: str, days: int) -> str:
    dt = datetime.datetime.strptime(date_str, "%Y-%m-%d")
    return (dt + datetime.timedelta(days=int(days))).strftime("%Y-%m-%d")

def _set_reminder(date_str: str, content: str) -> str:
    msg = f"[REMINDER SET] Date: {date_str} | Content: {content}"
    print(msg)
    return msg


# =============================================================================
# 2. TOOL REGISTRY  —  single source of truth
# =============================================================================

TOOLS: list[dict] = [
    {
        "name":        "calculator",
        "fn":          _calculator,
        "description": "Evaluate a plain math expression and return the result.",
        "args":        {"expression": 'str — math expression e.g. "10+10"'},
        "no_ref":      {"expression"},
    },
    {
        "name":        "get_current_time",
        "fn":          _get_current_time,
        "description": "Return the current local time as HH:MM:SS.",
        "args":        {},
    },
    {
        "name":        "add_hours",
        "fn":          _add_hours,
        "description": "Add N hours to a HH:MM:SS time string and return the new time.",
        "args":        {
            "time_str": "str — starting time in HH:MM:SS",
            "hours":    "float — hours to add",
        },
    },
    {
        "name":        "add_duration_to_datetime",
        "fn":          _add_duration_to_datetime,
        "description": "Add N days to a YYYY-MM-DD date and return the new date.",
        "args":        {
            "date_str": "str — starting date in YYYY-MM-DD",
            "days":     "int — days to add",
        },
    },
    {
        "name":        "set_reminder",
        "fn":          _set_reminder,
        "description": "Set a reminder on a given date. Content must never be empty.",
        "args":        {
            "date_str": "str — reminder date in YYYY-MM-DD",
            "content":  "str — reminder message, infer from context if not given",
        },
    },
    # ── Batch meta-tool ──────────────────────────────────────────────────────
    # fn=None because execute_plan intercepts it before reaching run_tool.
    # Still lives in TOOLS so the planner prompt and validator include it.
    {
        "name":        "batch",
        "fn":          None,
        "description": (
            "Run multiple INDEPENDENT tool calls in parallel. "
            "Use this when two or more calls have no dependency on each other."
        ),
        "args":        {
            "invocations": 'list — each item: {"name": "<tool_name>", "args": {...}}',
        },
    },
]

_TOOL_BY_NAME: dict[str, dict] = {t["name"]: t for t in TOOLS}


# =============================================================================
# 3. EXECUTION
# =============================================================================

def run_tool(name: str, args: dict) -> str:
    """Dispatch a single named tool. batch is NOT handled here."""
    tool = _TOOL_BY_NAME.get(name)
    if tool is None:
        raise ValueError(f"Unknown tool: {name!r}. Available: {list(_TOOL_BY_NAME)}")
    if tool["fn"] is None:
        raise ValueError(f"Tool {name!r} must be handled by the executor, not run_tool.")
    return str(tool["fn"](**args))


def run_batch(invocations: list) -> str:
    """
    Fan out a list of independent tool calls and return a JSON summary.
    Equivalent to the transcript's run_batch() that iterates invocations
    and calls run_tool on each one.
    """
    batch_output = []
    for inv in invocations:
        name = inv.get("name", "")
        args = inv.get("args", {})
        if isinstance(args, str):
            try:
                args = json.loads(args)
            except json.JSONDecodeError:
                args = {}
        print(f"    [batch] → {name}({args})")
        try:
            result = run_tool(name, args)
        except Exception as e:
            result = f"ERROR: {e}"
        print(f"    [batch] ← {result}")
        batch_output.append({"tool": name, "output": result})
    return json.dumps(batch_output)


def execute_plan(plan: list) -> dict:
    results = {}
    for step in plan:
        step_id = int(step["step"])
        tool    = step["tool"]
        args    = dict(step.get("args", {}))

        # Resolve $step_N references
        for key, val in args.items():
            if isinstance(val, str) and val.startswith("$step_"):
                ref = int(val.replace("$step_", ""))
                if ref not in results:
                    raise ValueError(
                        f"Step {step_id} references $step_{ref} "
                        "which hasn't run yet or was removed."
                    )
                args[key] = results[ref]

        # Intercept batch before it reaches run_tool
        if tool == "batch":
            invocations = args.get("invocations", [])
            print(f"  Step {step_id} → batch ({len(invocations)} parallel calls)")
            output = run_batch(invocations)
        else:
            print(f"  Step {step_id} → {tool}({args})")
            output = run_tool(tool, args)

        print(f"           ← {output}")
        results[step_id] = output

    return results


# =============================================================================
# 4. PLANNER  —  prompt generated from TOOLS, never written by hand
# =============================================================================

def _build_tool_docs() -> str:
    lines = []
    for t in TOOLS:
        arg_str = (
            json.dumps({k: f"<{v}>" for k, v in t["args"].items()})
            if t["args"] else "{}"
        )
        lines.append(f'- {t["name"]:32s} {t["description"]}')
        lines.append(f'  args: {arg_str}')
        if t.get("no_ref"):
            lines.append(f'  WARNING: These args must NOT be "$step_N" refs: {sorted(t["no_ref"])}')
    return "\n".join(lines)


def _build_planner_system() -> str:
    return f"""You are a planner. Convert user requests into a minimal JSON execution plan.

TOOLS
-----
{_build_tool_docs()}

RULES
-----
1. Return ONLY a valid JSON array — no explanation, no markdown, no backticks.
2. Each element: {{"step": <int>, "tool": "<name>", "args": {{...}}}}
3. To forward one step's output to a later step, use "$step_N" as the arg value.
4. Never pass "$step_N" into an arg marked WARNING above.
5. Use the minimum number of steps. Prefer dedicated tools over calculator for non-math tasks.
6. set_reminder content must never be empty — infer from context.
7. When two or more calls are INDEPENDENT (no output dependency), wrap them in a single batch step.

EXAMPLES
--------
User: What is 10+10?
[{{"step":1,"tool":"calculator","args":{{"expression":"10+10"}}}}]

User: What time is it now and what will it be 2 hours from now?
[{{"step":1,"tool":"get_current_time","args":{{}}}},{{"step":2,"tool":"add_hours","args":{{"time_str":"$step_1","hours":2}}}}]

User: Set a reminder 177 days after 2050-01-01.
[{{"step":1,"tool":"add_duration_to_datetime","args":{{"date_str":"2050-01-01","days":177}}}},{{"step":2,"tool":"set_reminder","args":{{"date_str":"$step_1","content":"Scheduled reminder"}}}}]

User: Set two reminders on 2050-06-27: one for doctor's appointment and one for taxes due.
[{{"step":1,"tool":"batch","args":{{"invocations":[{{"name":"set_reminder","args":{{"date_str":"2050-06-27","content":"Doctor's appointment"}}}},{{"name":"set_reminder","args":{{"date_str":"2050-06-27","content":"Taxes due"}}}}]}}}}]
"""


def _validate_plan(plan: list) -> list:
    """Drop steps with unknown tools or illegal $step_N references, then renumber."""
    cleaned = []
    for step in plan:
        tool_name = step.get("tool", "")
        tool_def  = _TOOL_BY_NAME.get(tool_name)
        args      = step.get("args", {})

        if tool_def is None:
            print(f"  [validator] Dropping unknown tool: {tool_name!r}")
            continue

        skip = False
        for arg_name in tool_def.get("no_ref", set()):
            val = args.get(arg_name, "")
            if isinstance(val, str) and "$step_" in val:
                print(f"  [validator] Dropping illegal $step ref in {arg_name!r}: {val!r}")
                skip = True
                break
        if not skip:
            cleaned.append(step)

    for i, step in enumerate(cleaned, start=1):
        step["step"] = i

    return cleaned


def make_plan(user_input: str, model: str) -> list:
    messages = [
        {"role": "system", "content": _build_planner_system()},
        {"role": "user",   "content": user_input},
    ]
    raw   = chat_llm(messages, model=model)
    clean = raw.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()
    try:
        plan = json.loads(clean)
        assert isinstance(plan, list)
        return _validate_plan(plan)
    except (json.JSONDecodeError, AssertionError):
        raise ValueError(f"Planner returned invalid JSON:\n{raw}")


# =============================================================================
# 5. LLM CALL
# =============================================================================

def chat_llm(messages: list, model: str = "llama3.1:8b") -> str:
    res = requests.post(
        "http://localhost:11434/api/chat",
        json={"model": model, "messages": messages, "stream": False},
    )
    res.raise_for_status()
    return res.json()["message"]["content"]


# =============================================================================
# 6. SUMMARISER
# =============================================================================

_SUMMARISER_SYSTEM = (
    "You are a helpful assistant. Given a user request and tool results, "
    "write a clear, concise answer in plain English. No JSON, no code."
)

def summarise(user_input: str, plan: list, results: dict, model: str) -> str:
    steps_text = "\n".join(
        f"  Step {s['step']} ({s['tool']}) -> {results.get(s['step'], '?')}"
        for s in plan
    )
    messages = [
        {"role": "system", "content": _SUMMARISER_SYSTEM},
        {"role": "user",   "content": f"User asked: {user_input}\n\nTool results:\n{steps_text}"},
    ]
    return chat_llm(messages, model=model)


# =============================================================================
# 7. MAIN
# =============================================================================

def run_conversation(user_input: str, model: str = "llama3.1:8b"):
    print(f"\n{'='*50}")
    print(f"USER: {user_input}")
    print(f"{'='*50}")

    try:
        plan = make_plan(user_input, model)
    except ValueError as e:
        print(f"Planning failed: {e}")
        return None
    print(f"\nPLAN: {json.dumps(plan, indent=2)}")

    print("\nEXECUTION:")
    try:
        results = execute_plan(plan)
    except Exception as e:
        print(f"Execution failed: {e}")
        return None

    final = summarise(user_input, plan, results, model)
    print(f"\nFINAL ANSWER: {final}")
    return final


if __name__ == "__main__":
    run_conversation("What is 10+10?")
    run_conversation("What time is it now and what will it be 2 hours from now?")
    run_conversation("Please set a reminder that is 177 days after January 1st, 2050.")
    # Batch test — mirrors the transcript's parallel reminders demo
    run_conversation(
        "Set two reminders on 2050-06-27: "
        "one for my doctor's appointment and one for taxes due."
    )
````

## Tools for structured data

````python
"""
structured_data.py
------------------
Translates "Tools for Structured Data" transcript to Ollama/Llama.

Key idea (from transcript):
  Instead of prompt engineering + stop sequences to get JSON,
  define a tool whose INPUT SCHEMA is exactly the structure you want.
  Force the LLM to call that tool via tool_choice.
  The tool's "input" block IS your structured output — no tool result needed.

Ollama differences vs Claude API:
  - Claude has tool_choice={"type":"tool","name":"..."}
    Ollama has no native tool_choice, so we enforce it via the system prompt.
  - Claude returns tool_calls inside the message.
    Ollama returns tool_calls as a list on the message object.
  - Both return tool args as a dict (or JSON string — we normalise).
"""

import json
import requests

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "llama3.1:8b"


# =============================================================================
# 1. SCHEMA BUILDER
#    In the transcript, schemas are hand-written JSON schema dicts.
#    We make a small helper so adding new schemas stays clean.
# =============================================================================

def make_schema(name: str, description: str, properties: dict, required: list = None) -> dict:
    """
    Build a tool schema dict — the Ollama/OpenAI function-calling format.

    properties: {field_name: {"type": "...", "description": "..."}}
    required:   list of required field names (defaults to all)

    This mirrors what the transcript calls "JSON schema spec for a tool
    whose inputs are going to be the structure of data we are looking for."
    """
    return {
        "type": "function",
        "function": {
            "name": name,
            "description": description,
            "parameters": {
                "type": "object",
                "properties": properties,
                "required": required if required is not None else list(properties.keys()),
            },
        },
    }


# =============================================================================
# 2. SCHEMAS
#    Each schema defines the exact JSON shape we want extracted.
#    Adding a new extraction target = one new schema + one call to chat().
# =============================================================================

# ── Article summary (matches transcript demo) ─────────────────────────────
article_summary_schema = make_schema(
    name        = "article_summary",
    description = "Extract structured metadata from a scholarly article.",
    properties  = {
        "title": {
            "type":        "string",
            "description": "Title of the article.",
        },
        "author": {
            "type":        "string",
            "description": "Author name.",
        },
        "key_insights": {
            "type":        "array",
            "items":       {"type": "string"},
            "description": "List of key insights or takeaways from the article.",
        },
    },
)

# ── Financial statement (second example from transcript) ──────────────────
financial_analysis_schema = make_schema(
    name        = "analyze_financial_statement",
    description = "Extract structured data from a financial statement.",
    properties  = {
        "balance": {
            "type":        "integer",
            "description": "Net balance as an integer.",
        },
        "key_insights": {
            "type":        "array",
            "items":       {"type": "string"},
            "description": "List of key financial insights.",
        },
    },
)


# =============================================================================
# 3. CHAT WITH FORCED TOOL CALL
#    Equivalent to Claude's tool_choice={"type":"tool","name":"<name>"}
#
#    Ollama has no native tool_choice parameter, so we:
#      1. Pass the schema in the tools list (so the LLM knows it exists)
#      2. Add a system prompt instruction that forces it to call that tool
#
#    This closely matches the transcript's intent: guarantee the LLM always
#    calls your specific extraction tool rather than answering in plain text.
# =============================================================================

def chat(
    messages:    list,
    tools:       list  = None,
    tool_choice: str   = None,   # name of the tool to force, e.g. "article_summary"
    model:       str   = MODEL,
) -> dict:
    """
    Send messages to Ollama and return the raw response message dict.
    Injects a tool_choice enforcement instruction when tool_choice is set.
    """
    msgs = list(messages)  # don't mutate caller's list

    # ── Enforce tool_choice via system prompt ─────────────────────────────
    # Claude API: tool_choice={"type":"tool","name":"article_summary"}
    # Ollama:     we prepend a system message that instructs the same behaviour
    if tool_choice and tools:
        tool_names = [t["function"]["name"] for t in tools]
        enforcement = (
            f"You MUST respond by calling the '{tool_choice}' tool. "
            f"Do NOT answer in plain text. "
            f"Available tools: {tool_names}. "
            f"Always call '{tool_choice}' with the correct arguments."
        )
        # Prepend so it doesn't get buried
        msgs = [{"role": "system", "content": enforcement}] + msgs

    payload = {"model": model, "messages": msgs, "stream": False}
    if tools:
        payload["tools"] = tools

    res = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]


# =============================================================================
# 4. STRUCTURED EXTRACTION
#    The transcript's core flow:
#      user text → chat(tools=[schema], tool_choice=name) → parse tool input
#    We stop after receiving the tool_call — no tool result sent back.
#    The tool's input IS the structured output we wanted.
# =============================================================================

def extract(text: str, schema: dict) -> dict | None:
    """
    Extract structured JSON from `text` using `schema` as the output shape.

    Returns the parsed dict on success, None on failure.

    Mirrors transcript: "once we get the assistant message back we say
    thanks for the JSON, we're done, we make no more follow-up requests."
    """
    tool_name = schema["function"]["name"]

    messages = [
        {"role": "user", "content": text},
    ]

    response = chat(messages, tools=[schema], tool_choice=tool_name)

    # ── Parse tool call from response ────────────────────────────────────
    tool_calls = response.get("tool_calls", [])
    if not tool_calls:
        print(f"⚠ Model did not call '{tool_name}'. Raw response:")
        print(response.get("content", "(no content)"))
        return None

    call = tool_calls[0]
    args = call["function"].get("arguments", {})

    # Ollama sometimes returns args as a JSON string — normalise
    if isinstance(args, str):
        try:
            args = json.loads(args)
        except json.JSONDecodeError as e:
            print(f"❌ Failed to parse tool arguments: {e}")
            print(f"Raw args: {args}")
            return None

    return args


# =============================================================================
# 5. DEMO
# =============================================================================

# ── Step 1: generate an article (mirrors transcript's first cell) ──────────
def generate_article(topic: str) -> str:
    """Ask the LLM to write a short scholarly article, return the text."""
    messages = [{"role": "user", "content":
        f"Write a one-paragraph scholarly article about {topic}. "
        f"Include a title and author name at the top."
    }]
    response = chat(messages)
    return response.get("content", "")


if __name__ == "__main__":

    # ── Demo 1: Article summary extraction ───────────────────────────────
    print("=" * 55)
    print("DEMO 1 — Article Summary Extraction")
    print("=" * 55)

    print("\n[Step 1] Generating article...")
    article = generate_article("distributed systems and fault tolerance")
    print(f"\nARTICLE:\n{article}")

    print("\n[Step 2] Extracting structured data...")
    result = extract(article, article_summary_schema)

    if result:
        print("\nEXTRACTED JSON:")
        print(json.dumps(result, indent=2))
        # Access individual fields — matches transcript's
        # "response.content[0].input" equivalent
        print(f"\n  title:        {result.get('title')}")
        print(f"  author:       {result.get('author')}")
        print(f"  key_insights: {result.get('key_insights')}")

    # ── Demo 2: Financial statement extraction ────────────────────────────
    print("\n" + "=" * 55)
    print("DEMO 2 — Financial Statement Extraction")
    print("=" * 55)

    financial_text = """
    Q3 Financial Report — Acme Corp
    Total revenue: $4.2M. Operating costs: $3.1M. Net balance: $1,100,000.
    Key highlights: revenue grew 12% YoY, cloud segment outperformed targets,
    headcount costs rose 8%, R&D investment increased by 20%.
    """

    print(f"\nINPUT TEXT:\n{financial_text.strip()}")
    print("\n[Extracting structured data...]")
    result = extract(financial_text, financial_analysis_schema)

    if result:
        print("\nEXTRACTED JSON:")
        print(json.dumps(result, indent=2))
````

# Fine grained tool calling

````python
"""
tool_streaming.py
-----------------
Translates "Fine Grained Tool Calling" transcript to Ollama/Llama.

TRANSCRIPT PART 1 — Basic tool streaming:
  Claude API: stream=True, handle ContentBlockDelta + InputJSONEvent chunks
  Ollama:     stream=True, handle text chunks + collect tool call at the end

TRANSCRIPT PART 2 — Fine-grained tool calling:
  Claude API: fine_grained=True disables server-side JSON validation.
              Tool arg chunks arrive immediately instead of buffered per top-level key.
              Risk: may receive invalid JSON — must handle errors yourself.
  Ollama:     No equivalent server-side validation exists.
              We simulate the same INTENT by:
                - streaming the raw response
                - doing incremental JSON parsing on each chunk ourselves
                - acting on partial data as early as possible
                - catching and handling invalid JSON explicitly

Tool used in demo (matches transcript):
  save_article — expects:
    abstract: str
    meta: {word_count: int, review: str}
  This nested structure is what makes the buffering/validation story interesting.
"""

import json
import requests

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "llama3.1:8b"


# =============================================================================
# 1. TOOL SCHEMA
#    Matches the transcript's save_article tool with nested meta object.
#    The nested structure (abstract + meta) mirrors the two-top-level-keys
#    example used to explain buffering behaviour in the transcript.
# =============================================================================

save_article_schema = {
    "type": "function",
    "function": {
        "name": "save_article",
        "description": "Save a generated scholarly article with metadata.",
        "parameters": {
            "type": "object",
            "properties": {
                "abstract": {
                    "type":        "string",
                    "description": "A short summary of the article.",
                },
                "meta": {
                    "type":        "object",
                    "description": "Metadata about the article.",
                    "properties": {
                        "word_count": {
                            "type":        "integer",
                            "description": "Approximate word count of the article.",
                        },
                        "review": {
                            "type":        "string",
                            "description": "A brief critical review of the article.",
                        },
                    },
                    "required": ["word_count", "review"],
                },
            },
            "required": ["abstract", "meta"],
        },
    },
}


# =============================================================================
# 2. PART 1 — Basic streaming chat
#
#    Transcript: "for every chunk in the stream, process it depending on type.
#                 ContentBlockDelta → show text. InputJSONEvent → print it."
#
#    Ollama equivalent:
#      - stream=True → yields newline-delimited JSON objects
#      - Each object has message.content (text chunk) while streaming
#      - Final object has message.tool_calls (complete tool call)
#      - We print text chunks live, collect tool call at the end
# =============================================================================

def chat_stream(messages: list, tools: list = None) -> dict | None:
    """
    Stream a response from Ollama. Print text chunks live as they arrive.
    Return the tool_call dict if the model made one, else None.

    Transcript parallel:
      text chunk   → ContentBlockDelta  (printed live)
      tool call    → InputJSONEvent     (collected, printed at end)
    """
    payload = {
        "model":    MODEL,
        "messages": messages,
        "stream":   True,       # ← enables streaming
    }
    if tools:
        payload["tools"] = tools

    print("\n[STREAMING] ", end="", flush=True)

    tool_call_result = None

    with requests.post(OLLAMA_URL, json=payload, stream=True) as res:
        res.raise_for_status()
        for raw_line in res.iter_lines():
            if not raw_line:
                continue

            chunk = json.loads(raw_line)
            msg   = chunk.get("message", {})

            # ── Text chunk → print immediately (ContentBlockDelta equivalent)
            text = msg.get("content", "")
            if text:
                print(text, end="", flush=True)

            # ── Tool call → arrives complete on the final chunk (Ollama behaviour)
            tool_calls = msg.get("tool_calls")
            if tool_calls:
                tool_call_result = tool_calls[0]

            # ── Stream finished
            if chunk.get("done"):
                print()  # newline after streamed text
                break

    return tool_call_result


# =============================================================================
# 3. PART 2 — Fine-grained equivalent: incremental JSON parsing
#
#    Transcript: "Fine Grained Tool Calling disables server-side validation.
#                 Chunks arrive immediately. You may get invalid JSON.
#                 You can act on partial data (e.g. word_count) much sooner."
#
#    Ollama reality: tool args always arrive complete (no chunk-by-chunk args).
#    Our simulation: once we have the args string, we parse it incrementally
#    key by key — acting on each field as soon as we find it, before the full
#    object is confirmed valid. This mirrors the transcript's intent:
#    "process parts of the tool call as quickly as possible."
#
#    We also explicitly handle invalid JSON — matching the transcript's
#    warning: "your code should assume you might be given invalid JSON."
# =============================================================================

def parse_args_incrementally(raw_args: str) -> None:
    """
    Simulate fine-grained processing: act on each top-level key as soon
    as it can be extracted, without waiting for the full object to be valid.

    Transcript parallel:
      - "abstract" key-value available → act on it immediately
      - "meta" key-value available → act on it immediately
      - Don't wait for the whole object before starting work
    """
    print("\n[FINE-GRAINED PARSING] Processing args incrementally...")

    # ── Step 1: try full parse first (happy path)
    try:
        full = json.loads(raw_args)
        print("  ✓ Full JSON valid — processing all keys:")
        for key, val in full.items():
            print(f"    [{key}] → {val}")
        return
    except json.JSONDecodeError:
        print("  ⚠ Full JSON invalid — falling back to key-by-key extraction")

    # ── Step 2: extract top-level keys one at a time (fine-grained fallback)
    # This mirrors what Claude API does internally during buffered validation:
    # validate and release one top-level key-value pair at a time.
    import re
    # Match simple string values
    for match in re.finditer(r'"(\w+)"\s*:\s*"((?:[^"\\]|\\.)*)"', raw_args):
        key, val = match.group(1), match.group(2)
        print(f"  [partial] string key found → {key}: {val[:60]}...")

    # Match numeric values
    for match in re.finditer(r'"(\w+)"\s*:\s*(\d+)', raw_args):
        key, val = match.group(1), match.group(2)
        print(f"  [partial] number key found → {key}: {val}")

    print("  ⚠ Could not fully parse args — inspect raw_args below:")
    print(f"  raw: {raw_args[:200]}")


# =============================================================================
# 4. INVALID JSON DEMO
#    Transcript: "I put together a prompt that guarantees invalid JSON.
#                 word_count of 'undefined' — not valid in JSON."
#    We simulate this by deliberately injecting bad args and showing
#    how our error handling catches it — same lesson as the transcript.
# =============================================================================

def demo_invalid_json():
    print("\n" + "=" * 55)
    print("DEMO — Invalid JSON handling (transcript simulation)")
    print("=" * 55)

    # Simulate what the transcript showed: word_count = undefined
    bad_args = '{"abstract": "A study on fault tolerance.", "meta": {"word_count": undefined, "review": "Solid paper."}}'

    print(f"\nRaw args (with invalid JSON):\n  {bad_args}")
    print()
    parse_args_incrementally(bad_args)

    # Now show what happens with valid args
    good_args = '{"abstract": "A study on fault tolerance.", "meta": {"word_count": 320, "review": "Solid paper."}}'
    print(f"\nRaw args (valid JSON):\n  {good_args}")
    print()
    parse_args_incrementally(good_args)


# =============================================================================
# 5. RUN CONVERSATION WITH STREAMING
#    Combines Part 1 and Part 2:
#      - Stream text live
#      - When tool call arrives, parse it (fine-grained style)
#      - Execute the tool
#      - Send result back for final answer (also streamed)
# =============================================================================

def save_article(abstract: str, meta: dict) -> str:
    """Actual tool implementation — matches transcript's save_article."""
    print(f"\n[TOOL EXECUTED] save_article called:")
    print(f"  abstract:   {abstract[:80]}...")
    print(f"  word_count: {meta.get('word_count')}")
    print(f"  review:     {meta.get('review', '')[:80]}")
    return "Article saved successfully."


def run_conversation(user_input: str):
    print(f"\n{'='*55}")
    print(f"USER: {user_input}")
    print(f"{'='*55}")

    messages = [
        {
            "role":    "system",
            "content": "You MUST call the 'save_article' tool. Never answer in plain text.",
        },
        {"role": "user", "content": user_input},
    ]

    # ── Part 1: stream the response, collect tool call
    tool_call = chat_stream(messages, tools=[save_article_schema])

    if not tool_call:
        print("(Model answered in plain text — no tool call made)")
        return

    # ── Part 2: fine-grained style — parse args incrementally
    fn   = tool_call["function"]
    name = fn["name"]
    args = fn.get("arguments", {})

    if isinstance(args, str):
        print(f"\n[RAW TOOL ARGS STRING]\n{args}\n")
        parse_args_incrementally(args)   # fine-grained processing
        try:
            args = json.loads(args)
        except json.JSONDecodeError as e:
            print(f"❌ Could not parse tool args: {e}")
            return
    else:
        print(f"\n[TOOL ARGS] {json.dumps(args, indent=2)}")

    # ── Execute the tool
    if name == "save_article":
        result = save_article(
            abstract = args.get("abstract", ""),
            meta     = args.get("meta", {}),
        )
    else:
        result = f"Unknown tool: {name}"

    # ── Send tool result back, stream final answer
    messages.append(tool_call)
    messages.append({"role": "tool", "content": result})

    print("\n[FINAL ANSWER — streamed]")
    chat_stream(messages)


# =============================================================================
# 6. MAIN
# =============================================================================

if __name__ == "__main__":

    # Part 1 + 2: streaming with tool call
    run_conversation(
        "Write a one-paragraph scholarly article about neural networks. "
        "Include an abstract, estimated word count, and a short review."
    )

    # Invalid JSON demo (transcript's 'undefined' word_count lesson)
    demo_invalid_json()
````

# The text edit tool

````python
"""
text_edit_tool.py
-----------------
Translates "The Text Edit Tool" transcript to Ollama/Llama.

Two sides of the tool:
  Schema         → tells the LLM what file operations are available
  Implementation → actual Python code that touches the filesystem

In Claude API: schema is a tiny stub that auto-expands server-side.
In Ollama:     we write the full schema ourselves.
"""

import json
import os
import re
import requests

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "llama3.1:8b"


# =============================================================================
# 1. TEXT EDITOR SCHEMA
# =============================================================================

text_editor_schema = {
    "type": "function",
    "function": {
        "name":        "str_replace_editor",
        "description": (
            "A text editor tool for viewing and modifying files. "
            "Use view to read a file or list a directory. "
            "Use str_replace to replace an exact string in a file. "
            "Use create to make a new file with content. "
            "Use undo_edit to undo the last change. "
            "ALWAYS provide the full file path in the path argument."
        ),
        "parameters": {
            "type":       "object",
            "properties": {
                "command": {
                    "type": "string",
                    "enum": ["view", "str_replace", "create", "undo_edit"],
                    "description": "The operation to perform.",
                },
                "path": {
                    "type":        "string",
                    "description": "Full path to the file or directory. Must never be empty.",
                },
                "old_str": {
                    "type":        "string",
                    "description": "Exact string to find and replace. Required for str_replace.",
                },
                "new_str": {
                    "type":        "string",
                    "description": "New string to insert in place of old_str. Required for str_replace.",
                },
                "file_text": {
                    "type":        "string",
                    "description": "Full content to write into the new file. Required for create.",
                },
            },
            "required": ["command", "path"],
        },
    },
}


# =============================================================================
# 2. TEXT EDITOR IMPLEMENTATION
# =============================================================================

class TextEditor:

    def __init__(self):
        self._history: list[tuple[str, str | None]] = []

    def view(self, path: str) -> str:
        if not os.path.exists(path):
            return f"ERROR: path not found: {path}"
        if os.path.isdir(path):
            entries = os.listdir(path)
            return f"Directory listing for {path}:\n" + "\n".join(entries)
        with open(path, "r", encoding="utf-8") as f:
            content = f.read()
        lines   = content.splitlines()
        numbered = "\n".join(f"{i+1:4d}  {line}" for i, line in enumerate(lines))
        return f"Contents of {path}:\n{numbered}"

    def str_replace(self, path: str, old_str: str, new_str: str) -> str:
        if not os.path.exists(path):
            return f"ERROR: file not found: {path}"
        if not old_str:
            return "ERROR: old_str is empty — cannot replace nothing."
        with open(path, "r", encoding="utf-8") as f:
            content = f.read()
        if old_str not in content:
            return f"ERROR: string not found in {path}."
        self._history.append((path, content))
        new_content = content.replace(old_str, new_str, 1)
        with open(path, "w", encoding="utf-8") as f:
            f.write(new_content)
        return f"Successfully replaced text in {path}."

    def create(self, path: str, file_text: str) -> str:
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as f:
                self._history.append((path, f.read()))
        else:
            self._history.append((path, None))
        parent = os.path.dirname(path)
        if parent:
            os.makedirs(parent, exist_ok=True)
        with open(path, "w", encoding="utf-8") as f:
            f.write(file_text)
        return f"Successfully created {path}."

    def undo_edit(self, path: str) -> str:
        if not self._history:
            return "ERROR: nothing to undo."
        last_path, old_content = self._history.pop()
        if old_content is None:
            os.remove(last_path)
            return f"Undone: deleted {last_path}."
        with open(last_path, "w", encoding="utf-8") as f:
            f.write(old_content)
        return f"Undone: restored {last_path}."

    def run(self, command: str, path: str, **kwargs) -> str:
        if command == "view":
            return self.view(path)
        elif command == "str_replace":
            return self.str_replace(path, kwargs.get("old_str", ""), kwargs.get("new_str", ""))
        elif command == "create":
            return self.create(path, kwargs.get("file_text", ""))
        elif command == "undo_edit":
            return self.undo_edit(path)
        else:
            return f"ERROR: unknown command: {command}"


# =============================================================================
# 3. PLAIN-TEXT LEAK RECOVERY
#    Small models sometimes print the tool call as text instead of using
#    the tool_calls field. We detect and recover from this.
# =============================================================================

def _extract_tool_call_from_text(text: str) -> dict | None:
    text = text.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()
    obj  = None
    try:
        obj = json.loads(text)
    except json.JSONDecodeError:
        match = re.search(r'\{.*\}', text, re.DOTALL)
        if match:
            try:
                obj = json.loads(match.group())
            except json.JSONDecodeError:
                pass
    if not obj or not obj.get("name"):
        return None
    args = obj.get("parameters") or obj.get("arguments") or {}
    return {"function": {"name": obj["name"], "arguments": args}}


# =============================================================================
# 4. LLM CALL
# =============================================================================

def chat(messages: list, tools: list = None) -> dict:
    payload = {"model": MODEL, "messages": messages, "stream": False}
    if tools:
        payload["tools"] = tools
    res = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]


# =============================================================================
# 5. RUN CONVERSATION
# =============================================================================

SYSTEM_PROMPT = """You are a coding assistant with access to a text editor tool called str_replace_editor.

RULES:
- Always use str_replace_editor to interact with files. Never describe what you would do — just do it.
- Always view a file before modifying it.
- Always provide the full file path in the path argument (e.g. "./main.py"). Never leave path empty.
- For str_replace: provide old_str as the exact existing code, and new_str as the replacement.
- For create: provide the full file content in file_text.
- Complete all requested tasks before giving a final answer.
"""

def run_conversation(user_input: str, editor: TextEditor):
    print(f"\n{'='*55}")
    print(f"USER: {user_input}")
    print(f"{'='*55}")

    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": user_input},
    ]

    max_iterations = 10
    iteration      = 0

    while iteration < max_iterations:
        iteration += 1
        response   = chat(messages, tools=[text_editor_schema])
        messages.append(response)

        tool_calls = response.get("tool_calls")

        # ── Plain-text leak fallback ──────────────────────────────────────
        if not tool_calls:
            content   = response.get("content", "")
            recovered = _extract_tool_call_from_text(content)
            if recovered:
                print("[recovered tool call from plain text]")
                tool_calls = [recovered]
                # Replace the appended message with a corrected one
                messages[-1] = {"role": "assistant", "tool_calls": tool_calls, "content": ""}

        # ── No tool call → LLM is done ────────────────────────────────────
        if not tool_calls:
            print(f"\nASSISTANT: {response.get('content', '')}")
            return

        # ── Execute tool calls ────────────────────────────────────────────
        for call in tool_calls:
            fn   = call["function"]
            name = fn["name"]
            args = fn.get("arguments", {})

            if isinstance(args, str):
                try:
                    args = json.loads(args)
                except json.JSONDecodeError:
                    args = {}

            command = args.get("command", "")
            path    = args.get("path",    "")

            print(f"\n[TOOL CALL] {name} → command={command!r}, path={path!r}")

            # ── Guard: reject empty path, ask LLM to retry ───────────────
            if not path:
                error_msg = (
                    "ERROR: path argument is empty. "
                    "Please provide the full file path (e.g. './main.py') and try again."
                )
                print(f"[TOOL RESULT] {error_msg}")
                messages.append({"role": "tool", "content": error_msg})
                continue

            result = editor.run(command, path, **{
                k: v for k, v in args.items() if k not in ("command", "path")
            })

            print(f"[TOOL RESULT]\n{result[:400]}")
            messages.append({"role": "tool", "content": result})

    print("⚠ Max iterations reached.")


# =============================================================================
# 6. DEMO
# =============================================================================

if __name__ == "__main__":

    # Setup: create a sample main.py
    with open("./main.py", "w") as f:
        f.write("def greeting():\n    print('hi there')\n")
    print("Created ./main.py for demo.")

    editor = TextEditor()

    # Demo 1: read and summarise
    run_conversation(
        "Open the ./main.py file and summarise its contents.",
        editor,
    )

    # Demo 2: add function + create test file
    run_conversation(
        "Open ./main.py and add a function to calculate pi to 5 decimal places. "
        "Then create a ./test.py file with tests for that function.",
        editor,
    )

    # Show results
    print("\n" + "=" * 55)
    print("FINAL FILE CONTENTS")
    print("=" * 55)
    for fname in ["./main.py", "./test.py"]:
        if os.path.exists(fname):
            print(f"\n--- {fname} ---")
            with open(fname) as f:
                print(f.read())
````

# The web search tool

````python
"""
web_search_tool.py
------------------
Translates "The Web Search Tool" transcript to Ollama/Llama.

Claude API:
  - Tiny schema stub → server auto-expands and runs the search
  - Returns structured blocks: text, search_result, citation
  - Supports max_uses and allowed_domains natively

Ollama translation:
  - We call DuckDuckGo ourselves (free, no API key needed)
  - We implement max_uses counter manually
  - We implement allowed_domains filtering manually
  - We build citation blocks ourselves from search results
  - LLM receives search results and formulates the answer
"""

import json
import requests
from duckduckgo_search import DDGS

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "llama3.1:8b"


# =============================================================================
# 1. WEB SEARCH SCHEMA
#    In Claude API this is a one-liner stub:
#      {"type": "web_search_20250305", "name": "web_search", "max_uses": 5}
#    ...which the server expands and fully manages.
#
#    In Ollama we write the full schema ourselves.
#    The LLM calls this tool, we execute the search, inject results back.
# =============================================================================

def make_web_search_schema(max_uses: int = 5, allowed_domains: list = None) -> dict:
    """
    Build the web search tool schema.
    Mirrors transcript's WebSearchSchema with max_uses and allowed_domains.

    max_uses        → how many times the LLM can search (transcript: capped at 5)
    allowed_domains → restrict search to specific domains (transcript: nih.gov example)
    """
    description = (
        "Search the web for up-to-date information to answer the user's question. "
        "Use this when you need current information or specialised knowledge. "
        f"You may search up to {max_uses} times total."
    )
    if allowed_domains:
        description += f" Only search within these domains: {', '.join(allowed_domains)}."

    return {
        "type": "function",
        "function": {
            "name":        "web_search",
            "description": description,
            "parameters": {
                "type":       "object",
                "properties": {
                    "query": {
                        "type":        "string",
                        "description": "The search query to look up.",
                    },
                },
                "required": ["query"],
            },
        },
        # Store config for use by the executor
        "_max_uses":        max_uses,
        "_allowed_domains": allowed_domains or [],
    }


# =============================================================================
# 2. SEARCH IMPLEMENTATION
#    This is the part Claude API handles invisibly.
#    We do it ourselves using DuckDuckGo.
#
#    Returns a structured result that mirrors Claude's WebSearchResultBlock:
#      title   → page title
#      url     → page URL
#      domain  → extracted domain (for citation rendering)
#      snippet → relevant text excerpt
# =============================================================================

def run_web_search(
    query:          str,
    allowed_domains: list = None,
    max_results:    int  = 5,
) -> list[dict]:
    """
    Execute a DuckDuckGo search and return structured results.
    Mirrors Claude's WebSearchResultBlock format.

    allowed_domains: if set, filter results to only these domains.
    This is the Ollama equivalent of Claude's server-side domain filtering.
    """
    print(f"\n  [SEARCH] query: {query!r}")
    if allowed_domains:
        print(f"  [SEARCH] allowed_domains: {allowed_domains}")

    results = []
    with DDGS() as ddgs:
        # If allowed_domains set, append site: operators to the query
        # e.g. "best leg exercises site:nih.gov"
        if allowed_domains:
            site_filter = " OR ".join(f"site:{d}" for d in allowed_domains)
            search_query = f"{query} {site_filter}"
        else:
            search_query = query

        for r in ddgs.text(search_query, max_results=max_results):
            url    = r.get("href", "")
            domain = url.split("/")[2] if url.startswith("http") else ""

            # Filter by allowed_domains if set
            if allowed_domains and not any(d in domain for d in allowed_domains):
                continue

            results.append({
                "title":   r.get("title",  ""),
                "url":     url,
                "domain":  domain,
                "snippet": r.get("body",   ""),
            })

    print(f"  [SEARCH] found {len(results)} result(s)")
    for r in results:
        print(f"    → {r['title'][:60]} ({r['domain']})")

    return results


def format_results_for_llm(results: list[dict]) -> str:
    """
    Format search results into a clear text block for the LLM.
    Mirrors what Claude's server injects back after a search.
    """
    if not results:
        return "No results found."

    lines = []
    for i, r in enumerate(results, 1):
        lines.append(f"[{i}] {r['title']}")
        lines.append(f"    URL: {r['url']}")
        lines.append(f"    {r['snippet']}")
        lines.append("")
    return "\n".join(lines)


# =============================================================================
# 3. CITATION RENDERER
#    Transcript: "render citation blocks so users can see the source"
#    We build this from the search results we collected during the session.
# =============================================================================

def render_citations(answer: str, sources: list[dict]) -> str:
    """
    Append a sources section to the answer.
    Mirrors the transcript's citation rendering pattern:
      domain | title | url | cited text
    """
    if not sources:
        return answer

    lines = [answer, "", "─" * 55, "SOURCES"]
    for i, s in enumerate(sources, 1):
        lines.append(f"\n[{i}] {s['title']}")
        lines.append(f"    Domain:  {s['domain']}")
        lines.append(f"    URL:     {s['url']}")
        if s.get("snippet"):
            lines.append(f"    Excerpt: {s['snippet'][:150]}...")

    return "\n".join(lines)


# =============================================================================
# 4. LLM CALL
# =============================================================================

def chat(messages: list, tools: list = None) -> dict:
    payload = {"model": MODEL, "messages": messages, "stream": False}
    if tools:
        # Strip private _ keys before sending to Ollama
        clean_tools = []
        for t in tools:
            clean = {k: v for k, v in t.items() if not k.startswith("_")}
            clean_tools.append(clean)
        payload["tools"] = clean_tools
    res = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]


# =============================================================================
# 5. RUN CONVERSATION
#    Multi-turn loop:
#      LLM decides to search → we run DuckDuckGo → inject results → LLM answers
#
#    max_uses counter mirrors transcript's "cap searches to 5"
#    allowed_domains mirrors transcript's "only search nih.gov"
# =============================================================================

def run_conversation(
    user_input:      str,
    allowed_domains: list = None,
    max_uses:        int  = 5,
):
    print(f"\n{'='*55}")
    print(f"USER: {user_input}")
    if allowed_domains:
        print(f"ALLOWED DOMAINS: {allowed_domains}")
    print(f"{'='*55}")

    schema   = make_web_search_schema(max_uses=max_uses, allowed_domains=allowed_domains)
    messages = [
        {
            "role":    "system",
            "content": (
                "You are a helpful assistant. "
                "Use the web_search tool to find up-to-date information "
                "before answering questions about current events or specialised topics. "
                "Always search before answering — do not rely on your training data alone."
            ),
        },
        {"role": "user", "content": user_input},
    ]

    search_count = 0     # tracks max_uses
    all_sources  = []    # accumulates sources for citation rendering

    max_iterations = 10
    for iteration in range(max_iterations):

        response   = chat(messages, tools=[schema])
        tool_calls = response.get("tool_calls")
        messages.append(response)

        # ── No tool call → LLM produced final answer ──────────────────────
        if not tool_calls:
            answer = response.get("content", "")
            # Render with citations
            final = render_citations(answer, all_sources)
            print(f"\nANSWER:\n{final}")
            return final

        # ── Tool call → execute the search ────────────────────────────────
        for call in tool_calls:
            fn    = call["function"]
            args  = fn.get("arguments", {})
            if isinstance(args, str):
                try:
                    args = json.loads(args)
                except json.JSONDecodeError:
                    args = {}

            query = args.get("query", "")

            # Enforce max_uses — mirrors transcript's cap
            if search_count >= max_uses:
                result_text = f"Search limit of {max_uses} reached. Cannot search further."
                print(f"\n  [SEARCH BLOCKED] {result_text}")
                messages.append({"role": "tool", "content": result_text})
                continue

            search_count += 1
            print(f"\n[SEARCH {search_count}/{max_uses}]")

            # Run the actual search
            results = run_web_search(
                query           = query,
                allowed_domains = allowed_domains,
                max_results     = 5,
            )

            # Accumulate sources for citation rendering
            all_sources.extend(results)

            # Inject results back into conversation
            result_text = format_results_for_llm(results)
            messages.append({"role": "tool", "content": result_text})

    print("⚠ Max iterations reached.")


# =============================================================================
# 6. DEMO — mirrors the transcript's two demonstrations
#
#   Demo 1: open web search (no domain restriction)
#   Demo 2: restricted to nih.gov (transcript's allowed_domains example)
# =============================================================================

if __name__ == "__main__":

    # Demo 1: general web search
    # Transcript: "ask Claude what's the best exercise for gaining leg muscle"
    run_conversation(
        "What are the best exercises for gaining leg muscle?",
        max_uses = 5,
    )

    # Demo 2: domain-restricted search
    # Transcript: "add allowed_domains = ['nih.gov'] to get scientific sources only"
    run_conversation(
        "What are the best exercises for gaining leg muscle?",
        allowed_domains = ["nih.gov"],
        max_uses        = 5,
    )
````

# Extended thinking

````python
"""
extended_thinking.py
--------------------
Translates "Extended Thinking" transcript to Ollama/Llama.

Two approaches:
  native    — deepseek-r1 outputs <think>...</think> automatically
  simulated — we prompt llama/qwen to reason inside tags, then parse
"""

import re
import requests

OLLAMA_URL = "http://localhost:11434/api/chat"


# =============================================================================
# 1. MODEL DETECTION
# =============================================================================

def list_local_models() -> list[str]:
    try:
        res = requests.get("http://localhost:11434/api/tags")
        res.raise_for_status()
        return [m["name"] for m in res.json().get("models", [])]
    except Exception:
        return []


def pick_thinking_model(preferred: str = None) -> tuple[str, str]:
    if preferred:
        approach = "native" if "deepseek" in preferred else "simulated"
        return preferred, approach
    available = list_local_models()
    for m in available:
        if "deepseek-r1" in m:
            return m, "native"
    for m in available:
        if "llama" in m or "qwen" in m:
            return m, "simulated"
    return "llama3.1:8b", "simulated"


# =============================================================================
# 2. TAG PARSER — shared by both approaches
#    Handles <think>, <thinking>, and cases where the tag appears mid-text
# =============================================================================

def parse_thinking_tags(content: str) -> tuple[str, str]:
    """
    Extract thinking and answer from model output.
    Tries <think> first (deepseek-r1), then <thinking> (simulated).
    Returns (thinking_text, answer_text).
    """
    for tag in ("think", "thinking"):
        pattern = rf"<{tag}>(.*?)</{tag}>"
        match   = re.search(pattern, content, re.DOTALL)
        if match:
            thinking = match.group(1).strip()
            # Answer is everything after the closing tag
            answer   = content[match.end():].strip()
            # Clean up any leftover "Your final answer:" prefix
            answer   = re.sub(r"(?i)^your final answer\s*:?\s*", "", answer).strip()
            return thinking, answer

    # No tags found — full response is the answer
    return "", content.strip()


# =============================================================================
# 3. APPROACH A — Native thinking (deepseek-r1)
# =============================================================================

def chat_native_thinking(messages: list, model: str) -> dict:
    payload = {"model": model, "messages": messages, "stream": False}
    res     = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    content  = res.json()["message"]["content"]
    thinking, answer = parse_thinking_tags(content)
    return {
        "thinking": thinking,
        "answer":   answer,
        "model":    model,
        "approach": "native",
        "redacted": False,
    }


# =============================================================================
# 4. APPROACH B — Simulated thinking (llama, qwen, etc.)
# =============================================================================

THINKING_SYSTEM_PROMPT = """You are a careful reasoning assistant. Always follow this exact format:

<thinking>
Write your complete step by step reasoning here.
Show all working. Think out loud fully. Do not skip steps.
</thinking>

Write your complete final answer here. This must be a full, standalone answer.
Do not write "Your final answer:" — just write the answer directly after the closing tag.

RULES:
- The <thinking> block MUST come first
- The answer MUST come after </thinking>
- Never put the answer inside the thinking block
- Never skip the thinking block"""


def chat_simulated_thinking(messages: list, model: str) -> dict:
    thinking_messages = [
        {"role": "system", "content": THINKING_SYSTEM_PROMPT},
    ] + messages

    payload = {"model": model, "messages": thinking_messages, "stream": False}
    res     = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    content  = res.json()["message"]["content"]
    thinking, answer = parse_thinking_tags(content)
    return {
        "thinking": thinking,
        "answer":   answer,
        "model":    model,
        "approach": "simulated",
        "redacted": False,
    }


# =============================================================================
# 5. UNIFIED chat() — mirrors transcript's chat(thinking=True/False)
# =============================================================================

def chat(
    messages:        list,
    thinking:        bool = False,
    thinking_budget: int  = 1024,
    model:           str  = None,
) -> dict:
    chosen_model, approach = pick_thinking_model(model)

    if not thinking:
        payload = {"model": chosen_model, "messages": messages, "stream": False}
        res     = requests.post(OLLAMA_URL, json=payload)
        res.raise_for_status()
        return {
            "thinking": "",
            "answer":   res.json()["message"]["content"],
            "model":    chosen_model,
            "approach": "none",
            "redacted": False,
        }

    if approach == "native":
        return chat_native_thinking(messages, chosen_model)
    else:
        return chat_simulated_thinking(messages, chosen_model)


# =============================================================================
# 6. RENDERER
# =============================================================================

def render_response(result: dict, show_thinking: bool = True) -> None:
    print(f"\nModel: {result['model']} ({result['approach']}) | Redacted: {result['redacted']}")
    if show_thinking and result["thinking"]:
        print("\n" + "─" * 55)
        print("THINKING BLOCK")
        print("─" * 55)
        print(result["thinking"])
    print("\n" + "─" * 55)
    print("ANSWER")
    print("─" * 55)
    print(result["answer"])


# =============================================================================
# 7. REDACTED THINKING SIMULATION
# =============================================================================

REDACTED_TEST_STRING = "anthropic_magic_string_trigger_redacted_thinking_LKJHF873"

def test_redacted_thinking(model: str = None) -> dict:
    """Send the magic string and treat missing thinking tags as redacted."""
    messages     = [{"role": "user", "content": REDACTED_TEST_STRING}]
    chosen_model, approach = pick_thinking_model(model)
    payload      = {"model": chosen_model, "messages": messages, "stream": False}
    res          = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    content      = res.json()["message"]["content"]
    has_thinking = bool(re.search(r"<think>|<thinking>", content, re.DOTALL))
    return {
        "thinking": "[REDACTED]" if not has_thinking else content,
        "answer":   content if not has_thinking else "",
        "model":    chosen_model,
        "approach": approach,
        "redacted": not has_thinking,
    }


# =============================================================================
# 8. DEMO
# =============================================================================

if __name__ == "__main__":

    # Demo 1: thinking OFF vs ON
    question = "Write a one paragraph guide to recursion."

    print("=" * 55)
    print("DEMO 1a — Thinking OFF")
    print("=" * 55)
    result = chat([{"role": "user", "content": question}], thinking=False)
    render_response(result, show_thinking=False)

    print("\n" + "=" * 55)
    print("DEMO 1b — Thinking ON")
    print("=" * 55)
    result = chat([{"role": "user", "content": question}], thinking=True)
    render_response(result, show_thinking=True)

    # Demo 2: complex reasoning
    print("\n" + "=" * 55)
    print("DEMO 2 — Complex reasoning (Bayes theorem)")
    print("=" * 55)
    result = chat(
        messages=[{"role": "user", "content":
            "I have 3 boxes. Box A has 2 red and 3 blue balls. "
            "Box B has 4 red and 1 blue. Box C has 1 red and 5 blue. "
            "I pick a box at random then draw a ball. It is red. "
            "What is the probability it came from Box B? Show full working."
        }],
        thinking=True,
        thinking_budget=2048,
    )
    render_response(result, show_thinking=True)

    # Demo 3: redacted thinking
    print("\n" + "=" * 55)
    print("DEMO 3 — Redacted thinking simulation")
    print("=" * 55)
    result = test_redacted_thinking()
    render_response(result, show_thinking=True)
    if result["redacted"]:
        print("\n[App handled redacted thinking gracefully — no crash]")
````

````python
"""
image_support.py
----------------
Translates "Vision / Image Support" transcript to Ollama/Llama.

Claude API:
  - Include an image block in the user message
  - image block accepts base64 data or a URL
  - Up to 100 images per request
  - Prompt engineering matters enormously for accuracy

Ollama translation:
  - Images encoded as base64 and sent in the "images" field
  - Model must be a vision model (llama3.2-vision or llava)
  - Same prompting lessons apply
"""

import base64
import requests
from pathlib import Path

OLLAMA_URL   = "http://localhost:11434/api/chat"
VISION_MODEL = "moondream"


# =============================================================================
# 1. IMAGE LOADING
# =============================================================================

def load_image_base64(path: str) -> str:
    """Read a local image file and return it as a base64 string."""
    with open(path, "rb") as f:
        return base64.standard_b64encode(f.read()).decode("utf-8")


def download_image(url: str, save_path: str) -> str:
    """
    Download an image from a URL to a local path.
    Uses a browser-like User-Agent to avoid 403 blocks.
    """
    print(f"  Downloading {url}...")
    headers = {"User-Agent": "Mozilla/5.0 (compatible; image-demo/1.0)"}
    res     = requests.get(url, headers=headers, timeout=15)
    res.raise_for_status()
    with open(save_path, "wb") as f:
        f.write(res.content)
    print(f"  Saved to {save_path}")
    return save_path


# =============================================================================
# 2. SAMPLE IMAGES
#    Using picsum.photos — free, reliable, no auth needed
# =============================================================================

SAMPLE_IMAGES = {
    "forest":  ("https://picsum.photos/id/10/640/480",  "/tmp/sample_forest.jpg"),
    "city":    ("https://picsum.photos/id/43/640/480",  "/tmp/sample_city.jpg"),
    "nature":  ("https://picsum.photos/id/15/640/480",  "/tmp/sample_nature.jpg"),
}

def get_sample_image(name: str) -> str:
    """Download a sample image if not already cached, return local path."""
    url, path = SAMPLE_IMAGES[name]
    if not Path(path).exists():
        download_image(url, path)
    else:
        print(f"  Using cached {path}")
    return path


# =============================================================================
# 3. MESSAGE BUILDERS
#    Ollama puts images in a separate "images" list alongside "content" text.
#
#    Claude API structure:
#      content = [{"type": "image", "source": {...}}, {"type": "text", ...}]
#
#    Ollama structure:
#      {"role": "user", "content": "prompt text", "images": [base64_string]}
# =============================================================================

def make_image_message(prompt: str, *image_paths: str) -> dict:
    """
    Build a user message with one or more images.
    Mirrors Claude's image block + text block pattern.
    """
    encoded = [load_image_base64(p) for p in image_paths]
    return {
        "role":    "user",
        "content": prompt,
        "images":  encoded,
    }


# =============================================================================
# 4. CHAT
# =============================================================================

def chat(messages: list, model: str = VISION_MODEL) -> str:
    payload = {"model": model, "messages": messages, "stream": False}
    res     = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]["content"]


# =============================================================================
# 5. DEMO
# =============================================================================

if __name__ == "__main__":

    # ── Demo 1: Simple vs detailed prompt ─────────────────────────────────
    # Transcript: "the number one way to get good results is prompting technique"
    print("=" * 55)
    print("DEMO 1 — Simple vs detailed prompt")
    print("=" * 55)

    path = get_sample_image("forest")

    # Simple prompt — transcript warns this often gives poor results
    print("\n[Simple prompt]")
    result = chat([make_image_message("What is in this image?", path)])
    print(result)

    # Detailed prompt with analysis steps
    # Transcript: "provide a series of steps to go through"
    print("\n[Detailed prompt with steps]")
    detailed_prompt = """Analyse this image carefully using the following steps:
1. Identify the main subject or objects.
2. Describe the colours, shapes, and spatial arrangement.
3. Note any text, labels, or symbols present.
4. Describe the overall context or setting.
5. Provide a one-sentence summary.

Walk through each step explicitly."""

    result = chat([make_image_message(detailed_prompt, path)])
    print(result)

    # ── Demo 2: Comparing two images ──────────────────────────────────────
    # Transcript: "we could ask Claude to compare different images"
    print("\n" + "=" * 55)
    print("DEMO 2 — Comparing two images")
    print("=" * 55)

    path1 = get_sample_image("forest")
    path2 = get_sample_image("city")

    prompt = """You are given two images. Compare them by:
1. Describing what each image shows.
2. Identifying the key differences between them.
3. Identifying anything they have in common.
4. Stating which image has more human activity and why."""

    result = chat([make_image_message(prompt, path1, path2)])
    print(result)

    # ── Demo 3: Few-shot prompting ─────────────────────────────────────────
    # Transcript: "alternate image and text blocks to provide examples"
    # We simulate with a multi-turn conversation
    print("\n" + "=" * 55)
    print("DEMO 3 — Few-shot prompting")
    print("=" * 55)

    path1 = get_sample_image("forest")
    path2 = get_sample_image("nature")

    # Show one example first, then ask about the second image
    # The model follows the established pattern
    messages = [
        make_image_message(
            "Describe this image in exactly one sentence starting with 'This image shows'.",
            path1
        ),
        {
            "role":    "assistant",
            "content": "This image shows a dense forest path surrounded by tall trees with dappled light filtering through the canopy.",
        },
        make_image_message(
            "Describe this image in exactly one sentence starting with 'This image shows'.",
            path2
        ),
    ]

    result = chat(messages)
    print(f"Few-shot result:\n{result}")

    # ── Demo 4: Structured assessment ─────────────────────────────────────
    # Mirrors transcript's fire risk assessment use case
    # Same idea: structured multi-step prompt → consistent output format
    print("\n" + "=" * 55)
    print("DEMO 4 — Structured assessment")
    print("=" * 55)

    path = get_sample_image("city")

    assessment_prompt = """You are conducting a visual safety assessment of this image.
Follow these steps exactly:

Step 1: Identify the main subject and setting.
Step 2: Identify any potential risks or hazards visible.
Step 3: Identify any positive or safe elements.
Step 4: Assign a risk rating from 1 (very safe) to 4 (high risk).
Step 5: Write a one-sentence summary with your final rating.

Provide a clearly labelled response for each step."""

    result = chat([make_image_message(assessment_prompt, path)])
    print(result)
````

# PDF support

````python
"""
pdf_support.py
--------------
Translates "PDF Support" transcript to Ollama/Llama.

Claude API:
  - Send PDF as base64 with type="document", media_type="application/pdf"
  - Claude natively reads text, tables, charts, images inside the PDF
  - Almost identical to image support, just different type fields

Ollama translation:
  - No native PDF parsing in most models
  - We extract text ourselves using pypdf
  - Send extracted text as a document block to the model
  - Model's job (summarise, answer questions) is identical

pip install pypdf
"""

import requests
from pathlib import Path
from pypdf import PdfReader

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "qwen2.5:7b"   # any text model works — no vision needed


# =============================================================================
# 1. PDF EXTRACTION
#    This is the step Claude API handles invisibly.
#    We do it ourselves using pypdf.
# =============================================================================

def extract_pdf_text(path: str) -> dict:
    """
    Extract text from a PDF file page by page.
    Returns:
      {
        "full_text":  str,            # all pages joined
        "pages":      [str, ...],     # text per page
        "page_count": int,
        "path":       str,
      }
    Mirrors what Claude receives internally when you send a PDF.
    """
    reader = PdfReader(path)
    pages  = []

    for i, page in enumerate(reader.pages):
        text = page.extract_text() or ""
        pages.append(text.strip())

    full_text = "\n\n--- Page Break ---\n\n".join(
        f"[Page {i+1}]\n{text}" for i, text in enumerate(pages) if text
    )

    return {
        "full_text":  full_text,
        "pages":      pages,
        "page_count": len(reader.pages),
        "path":       path,
    }


def pdf_summary(pdf: dict) -> str:
    """Print a quick summary of what was extracted."""
    print(f"  File:       {pdf['path']}")
    print(f"  Pages:      {pdf['page_count']}")
    print(f"  Characters: {len(pdf['full_text'])}")
    non_empty = sum(1 for p in pdf["pages"] if p.strip())
    print(f"  Non-empty pages: {non_empty}")


# =============================================================================
# 2. MESSAGE BUILDER
#    Mirrors Claude's document block structure.
#
#    Claude API:
#      {"type": "document", "source": {"type": "base64",
#       "media_type": "application/pdf", "data": "<base64>"}, "title": "..."}
#
#    Ollama equivalent:
#      We inject the extracted text as a clearly labelled document block
#      in the system or user message. The model treats it as a reference document.
# =============================================================================

def make_pdf_message(pdf: dict, prompt: str, title: str = None) -> list:
    """
    Build a message list with the PDF content injected as a document.
    Returns a messages list ready to pass to chat().

    Mirrors transcript's flow:
      "send the PDF content to Claude along with a question"
    """
    doc_title = title or Path(pdf["path"]).name

    # Inject document as a system message so the model treats it as reference
    system_content = (
        f"The following is the content of a document called '{doc_title}'. "
        f"Use it to answer the user's question accurately.\n\n"
        f"{'─' * 40}\n"
        f"{pdf['full_text']}\n"
        f"{'─' * 40}"
    )

    return [
        {"role": "system", "content": system_content},
        {"role": "user",   "content": prompt},
    ]


# =============================================================================
# 3. CHAT
# =============================================================================

def chat(messages: list, model: str = MODEL) -> str:
    payload = {"model": model, "messages": messages, "stream": False}
    res     = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]["content"]


# =============================================================================
# 4. SAMPLE PDF CREATOR
#    Since we can't attach earth.pdf from the transcript,
#    we create a sample PDF with known content to demo against.
#    pip install pypdf already installed reportlab if available,
#    otherwise we write a minimal PDF by hand.
# =============================================================================

def create_sample_pdf(path: str) -> str:
    """
    Create a sample PDF about Earth for testing.
    Mirrors the transcript's earth.pdf demo document.
    """
    # Minimal valid PDF written by hand — no extra libraries needed
    content = """\
Earth is the third planet from the Sun and the only astronomical object known to harbor life.
It formed approximately 4.5 billion years ago.

Earth's atmosphere and oceans were formed by volcanic activity and outgassing.
Water vapor from volcanoes condensed to form the oceans over millions of years.

The planet's surface is divided into several tectonic plates that move slowly over time.
This movement causes earthquakes, volcanic eruptions, and the formation of mountain ranges.

Earth has one natural satellite, the Moon, which stabilizes its axial tilt.
The Moon also drives ocean tides through gravitational interaction.

Key facts:
- Diameter: 12,742 km
- Mass: 5.97 x 10^24 kg
- Surface temperature: -88 to 58 degrees Celsius
- Atmosphere: 78% nitrogen, 21% oxygen, 1% other gases
- Age: approximately 4.54 billion years

Pages 4-5: Earth's atmosphere was originally rich in carbon dioxide and water vapor.
Over billions of years, photosynthetic organisms converted CO2 to oxygen,
gradually transforming the atmosphere into its current composition.
"""

    try:
        # Try reportlab if available
        from reportlab.pdfgen import canvas
        from reportlab.lib.pagesizes import A4

        c = canvas.Canvas(path, pagesize=A4)
        width, height = A4
        y = height - 50

        c.setFont("Helvetica-Bold", 14)
        c.drawString(50, y, "Earth — Reference Document")
        y -= 30

        c.setFont("Helvetica", 11)
        for line in content.strip().split("\n"):
            if y < 50:
                c.showPage()
                y = height - 50
                c.setFont("Helvetica", 11)
            c.drawString(50, y, line)
            y -= 16

        c.save()

    except ImportError:
        # reportlab not available — write a minimal PDF manually
        pdf_content = f"""%PDF-1.4
1 0 obj
<< /Type /Catalog /Pages 2 0 R >>
endobj

2 0 obj
<< /Type /Pages /Kids [3 0 R] /Count 1 >>
endobj

3 0 obj
<< /Type /Page /Parent 2 0 R /MediaBox [0 0 612 792]
   /Contents 4 0 R /Resources << /Font << /F1 5 0 R >> >> >>
endobj

4 0 obj
<< /Length 44 >>
stream
BT /F1 12 Tf 50 750 Td (Earth Reference Document) Tj ET
endstream
endobj

5 0 obj
<< /Type /Font /Subtype /Type1 /BaseFont /Helvetica >>
endobj

xref
0 6
0000000000 65535 f
0000000009 00000 n
0000000058 00000 n
0000000115 00000 n
0000000274 00000 n
0000000370 00000 n

trailer
<< /Size 6 /Root 1 0 R >>
startxref
441
%%EOF"""
        with open(path, "w") as f:
            f.write(pdf_content)

        # Also write a companion text file with full content
        txt_path = path.replace(".pdf", "_content.txt")
        with open(txt_path, "w") as f:
            f.write(content)
        print(f"  Note: reportlab not installed. Full text saved to {txt_path}")
        print(f"  For PDF demo, install: pip install reportlab")

    print(f"  Created sample PDF: {path}")
    return path


# =============================================================================
# 5. DEMO
# =============================================================================

if __name__ == "__main__":

    # Setup — create sample PDF
    pdf_path = "/tmp/earth.pdf"
    print("Creating sample PDF...")
    create_sample_pdf(pdf_path)

    # Extract text from PDF
    print("\nExtracting PDF content...")
    pdf = extract_pdf_text(pdf_path)
    pdf_summary(pdf)

    # ── Demo 1: Basic summarisation
    # Transcript: "ask Claude to summarise the document in one sentence"
    print("\n" + "=" * 55)
    print("DEMO 1 — Summarise in one sentence")
    print("=" * 55)

    messages = make_pdf_message(
        pdf,
        prompt = "Summarise this document in one sentence.",
        title  = "earth.pdf",
    )
    result = chat(messages)
    print(result)

    # ── Demo 2: Specific question
    # Transcript: "ask how Earth's atmosphere and oceans were formed"
    print("\n" + "=" * 55)
    print("DEMO 2 — Specific question from document")
    print("=" * 55)

    messages = make_pdf_message(
        pdf,
        prompt = "How were Earth's atmosphere and oceans formed?",
        title  = "earth.pdf",
    )
    result = chat(messages)
    print(result)

    # ── Demo 3: Key facts extraction
    # Transcript: "Claude can extract tables, figures, structured data"
    print("\n" + "=" * 55)
    print("DEMO 3 — Extract structured data")
    print("=" * 55)

    messages = make_pdf_message(
        pdf,
        prompt = (
            "Extract all the key facts from this document as a structured list. "
            "Include numbers, dates, and measurements where mentioned."
        ),
        title = "earth.pdf",
    )
    result = chat(messages)
    print(result)

    # ── Demo 4: Multi-turn conversation over the same PDF
    # Mirrors transcript's intent: keep asking questions about the same document
    print("\n" + "=" * 55)
    print("DEMO 4 — Multi-turn Q&A over the document")
    print("=" * 55)

    doc_title  = "earth.pdf"
    system_msg = {
        "role":    "system",
        "content": (
            f"The following is the content of '{doc_title}'. "
            f"Answer all questions based only on this document.\n\n"
            f"{'─'*40}\n{pdf['full_text']}\n{'─'*40}"
        ),
    }

    history = [system_msg]
    questions = [
        "What is the age of Earth?",
        "What percentage of Earth's atmosphere is oxygen?",
        "How does the Moon affect Earth?",
    ]

    for q in questions:
        print(f"\nQ: {q}")
        history.append({"role": "user", "content": q})
        answer = chat(history)
        history.append({"role": "assistant", "content": answer})
        print(f"A: {answer}")
````

# Citations

````python
"""
citations.py
------------
Translates "Citations" transcript to Ollama/Llama.

Claude API:
  - Add citations={"enabled": True} to a document block
  - Response contains citation_page_location (PDF) or citation_char_location (text)
  - Each citation has: cited_text, document_title, start_page/end_page or start_char/end_char
  - Allows building UI where users can verify sources

Ollama translation:
  - No native citation support
  - We prompt the model to cite sources explicitly using [n] notation
  - We parse the citations out of the response
  - We render them in the same format the transcript describes
  - Two modes matching transcript: PDF (page-based) and plain text (char-based)
"""

import re
import requests
from pypdf import PdfReader

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "qwen2.5:7b"


# =============================================================================
# 1. DOCUMENT LOADERS
#    Two sources matching the transcript:
#      load_pdf()       → citation_page_location equivalent
#      load_plaintext() → citation_char_location equivalent
# =============================================================================

def load_pdf(path: str, title: str = None) -> dict:
    """
    Load a PDF and return a document dict with page-level content.
    Mirrors Claude's PDF document block + page location tracking.
    """
    reader = PdfReader(path)
    pages  = []
    for i, page in enumerate(reader.pages):
        text = (page.extract_text() or "").strip()
        if text:
            pages.append({"page": i + 1, "text": text})

    return {
        "type":  "pdf",
        "title": title or path,
        "pages": pages,
        "full_text": "\n\n".join(
            f"[Page {p['page']}]\n{p['text']}" for p in pages
        ),
    }


def load_plaintext(text: str, title: str = "document") -> dict:
    """
    Load plain text and return a document dict with character positions.
    Mirrors Claude's plain text document block + char location tracking.
    """
    # Split into paragraphs for granular citation
    paragraphs = []
    pos = 0
    for para in re.split(r"\n{2,}", text.strip()):
        para = para.strip()
        if para:
            paragraphs.append({
                "text":       para,
                "start_char": pos,
                "end_char":   pos + len(para),
            })
            pos += len(para) + 2  # +2 for the \n\n separator

    return {
        "type":       "text",
        "title":      title,
        "paragraphs": paragraphs,
        "full_text":  text.strip(),
    }


# =============================================================================
# 2. CITATION PROMPT BUILDER
#    We instruct the model to cite sources using [n] markers.
#    Then we parse those markers and resolve them to actual source passages.
#
#    Transcript intent:
#      "citations allow Claude to refer to some outside source directly
#       and say it got its answer by looking at some other document"
# =============================================================================

CITATION_SYSTEM = """You are a precise research assistant. Answer questions based ONLY on the provided document.

CITATION RULES:
- Every factual claim MUST be supported by a citation
- Use [1], [2], [3] etc. inline to mark citations
- After your answer, list all citations in this exact format:

CITATIONS:
[1] "exact quoted text from document" (location)
[2] "exact quoted text from document" (location)

- Quote the exact words from the document — do not paraphrase in the citation
- Keep citations short — one or two sentences maximum
- If the document does not contain the answer, say so explicitly
"""

def make_citation_messages(doc: dict, prompt: str) -> list:
    """Build messages that instruct the model to cite its sources."""

    if doc["type"] == "pdf":
        doc_context = (
            f"DOCUMENT: {doc['title']}\n"
            f"{'─' * 40}\n"
            f"{doc['full_text']}\n"
            f"{'─' * 40}"
        )
    else:
        doc_context = (
            f"DOCUMENT: {doc['title']}\n"
            f"{'─' * 40}\n"
            f"{doc['full_text']}\n"
            f"{'─' * 40}"
        )

    return [
        {"role": "system", "content": CITATION_SYSTEM},
        {"role": "user",   "content": f"{doc_context}\n\nQuestion: {prompt}"},
    ]


# =============================================================================
# 3. CITATION PARSER
#    Parse the model's response into answer text + structured citation objects.
#
#    Transcript parallel:
#      citation_page_location → {cited_text, document_title, start_page, end_page}
#      citation_char_location → {cited_text, document_title, start_char, end_char}
#
#    We build the same structure from the model's [n] markers.
# =============================================================================

def parse_citations(response: str, doc: dict) -> dict:
    """
    Split model response into answer + citations list.
    Resolves cited text back to source locations in the document.

    Returns:
      {
        "answer":    str,
        "citations": [
          {
            "index":      int,
            "cited_text": str,
            "title":      str,
            "location":   str,   # "page X" or "chars X-Y"
            "start":      int,
            "end":        int,
          }
        ]
      }
    """
    # Split answer from citations section
    parts  = re.split(r"\n+CITATIONS:\s*\n", response, maxsplit=1)
    answer = parts[0].strip()
    raw_citations = parts[1].strip() if len(parts) > 1 else ""

    # Parse each citation line: [n] "quoted text" (location)
    citations = []
    pattern   = r'\[(\d+)\]\s*["\u201c]([^"\u201d]+)["\u201d]\s*(?:\(([^)]+)\))?'

    for match in re.finditer(pattern, raw_citations):
        index      = int(match.group(1))
        cited_text = match.group(2).strip()
        location   = match.group(3) or ""

        # Resolve location back to document
        resolved = resolve_location(cited_text, doc)

        citations.append({
            "index":      index,
            "cited_text": cited_text,
            "title":      doc["title"],
            "location":   resolved.get("location", location),
            "start":      resolved.get("start", 0),
            "end":        resolved.get("end",   0),
        })

    return {"answer": answer, "citations": citations}


def resolve_location(cited_text: str, doc: dict) -> dict:
    """
    Find where cited_text appears in the document.
    Returns page number (PDF) or char position (plain text).

    Mirrors transcript's citation_page_location and citation_char_location.
    """
    cited_lower = cited_text.lower()

    if doc["type"] == "pdf":
        for page in doc["pages"]:
            if cited_lower in page["text"].lower():
                return {
                    "location": f"page {page['page']}",
                    "start":    page["page"],
                    "end":      page["page"],
                }
        return {"location": "unknown page", "start": 0, "end": 0}

    else:  # plain text
        pos = doc["full_text"].lower().find(cited_lower)
        if pos >= 0:
            return {
                "location": f"chars {pos}–{pos + len(cited_text)}",
                "start":    pos,
                "end":      pos + len(cited_text),
            }
        return {"location": "unknown position", "start": 0, "end": 0}


# =============================================================================
# 4. CITATION RENDERER
#    Transcript: "render citations so users can see exactly where Claude
#                 is getting its information from"
#
#    We print in the same format as the transcript's popup UI:
#      source title | location | cited text
# =============================================================================

def render_response(parsed: dict) -> None:
    """
    Render answer with inline citation markers and a sources section.
    Mirrors the transcript's popup citation UI in terminal form.
    """
    print("\nANSWER:")
    print("─" * 55)
    print(parsed["answer"])

    if parsed["citations"]:
        print("\n" + "─" * 55)
        print("SOURCES:")
        for c in parsed["citations"]:
            print(f"\n  [{c['index']}] {c['title']}")
            print(f"       Location: {c['location']}")
            print(f"       Cited:    \"{c['cited_text'][:120]}\"")
    else:
        print("\n(No citations found in response)")


# =============================================================================
# 5. FULL PIPELINE
# =============================================================================

def chat(messages: list) -> str:
    payload = {"model": MODEL, "messages": messages, "stream": False}
    res     = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()
    return res.json()["message"]["content"]


def ask_with_citations(doc: dict, question: str) -> dict:
    """
    Full pipeline: question → cited answer → parsed citations.
    Mirrors transcript's flow: send document + question → get cited response.
    """
    messages = make_citation_messages(doc, question)
    response = chat(messages)
    return parse_citations(response, doc)


# =============================================================================
# 6. DEMO
# =============================================================================

EARTH_TEXT = """
Earth is the third planet from the Sun and the only astronomical object known to harbor life.
It formed approximately 4.5 billion years ago.

Earth's atmosphere and oceans were formed by volcanic activity and outgassing.
Water vapor from volcanoes condensed to form the oceans over millions of years.
The early atmosphere was rich in carbon dioxide, methane, and water vapor.

The planet's surface is divided into several tectonic plates that move slowly over time.
This movement causes earthquakes, volcanic eruptions, and the formation of mountain ranges.

Earth has one natural satellite, the Moon, which stabilizes its axial tilt.
The Moon also drives ocean tides through gravitational interaction with Earth.

Key facts about Earth:
- Diameter: 12,742 km
- Age: approximately 4.54 billion years
- Surface temperature range: -88 to 58 degrees Celsius
- Atmosphere composition: 78% nitrogen, 21% oxygen, 1% other gases
- One natural satellite: the Moon
"""


if __name__ == "__main__":

    # ── Demo 1: Plain text citations
    # Transcript: "you can cite from plain text — gives citation_char_location"
    print("=" * 55)
    print("DEMO 1 — Plain text citations (char location)")
    print("=" * 55)

    doc = load_plaintext(EARTH_TEXT, title="Earth Article")

    parsed = ask_with_citations(
        doc,
        "How were Earth's atmosphere and oceans formed?"
    )
    render_response(parsed)

    # ── Demo 2: Second question — different citations
    print("\n" + "=" * 55)
    print("DEMO 2 — Different question, different citations")
    print("=" * 55)

    parsed = ask_with_citations(
        doc,
        "What is the role of the Moon in relation to Earth?"
    )
    render_response(parsed)

    # ── Demo 3: PDF citations
    # Transcript: "citations work with PDF documents too — gives citation_page_location"
    print("\n" + "=" * 55)
    print("DEMO 3 — PDF citations (page location)")
    print("=" * 55)

    # Create a simple PDF from our earth text
    pdf_path = "/tmp/earth_citations.pdf"
    try:
        from reportlab.pdfgen import canvas
        from reportlab.lib.pagesizes import A4

        c      = canvas.Canvas(pdf_path, pagesize=A4)
        width, height = A4

        # Page 1
        c.setFont("Helvetica-Bold", 14)
        c.drawString(50, height - 50, "Earth — Reference Document")
        c.setFont("Helvetica", 11)
        y = height - 80
        page1_lines = EARTH_TEXT.strip().split("\n")[:10]
        for line in page1_lines:
            c.drawString(50, y, line.strip())
            y -= 18
        c.showPage()

        # Page 2
        c.setFont("Helvetica", 11)
        y = height - 50
        page2_lines = EARTH_TEXT.strip().split("\n")[10:]
        for line in page2_lines:
            c.drawString(50, y, line.strip())
            y -= 18
        c.save()

        doc_pdf = load_pdf(pdf_path, title="earth.pdf")
        parsed  = ask_with_citations(
            doc_pdf,
            "What are the key facts about Earth's size and age?"
        )
        render_response(parsed)

    except ImportError:
        print("reportlab not installed — skipping PDF demo.")
        print("Run: pip install reportlab")
        print("Plain text demo above covers the same citation concept.")

    # ── Demo 4: Question not in document
    # Tests that model correctly says "not in document" rather than hallucinating
    print("\n" + "=" * 55)
    print("DEMO 4 — Question outside document scope")
    print("=" * 55)

    doc = load_plaintext(EARTH_TEXT, title="Earth Article")
    parsed = ask_with_citations(
        doc,
        "What is the population of Earth?"
    )
    render_response(parsed)
````

# Prompt caching

````python
"""
prompt_caching.py
-----------------
Translates "Prompt Caching" transcript to Ollama/Llama.

Claude API:
  - Add cache_control={"type": "ephemeral"} to the last tool schema,
    system prompt, or message block you want to cache
  - Cache lives for 1 hour
  - Minimum 1024 tokens to cache
  - Up to 4 cache breakpoints per request
  - Response usage field shows: cache_creation_input_tokens, cache_read_input_tokens

Ollama reality:
  - No server-side caching exists
  - We simulate the LOGIC: track what would be cached, show hits/misses,
    count tokens, measure time savings
  - Code is structured identically to Claude API so switching is one line

Two modes in this file:
  MODE = "ollama"    → simulate caching logic, real LLM responses
  MODE = "claude"    → real caching (requires ANTHROPIC_API_KEY)
"""

import os
import time
import json
import hashlib
import requests

MODE       = "ollama"    # switch to "claude" when you have an API key
OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "qwen2.5:7b"


# =============================================================================
# 1. TOKEN COUNTER (approximation)
#    Claude API returns exact token counts in the usage field.
#    We approximate: 1 token ≈ 4 characters.
#    Good enough for understanding caching behaviour.
# =============================================================================

def count_tokens(text: str) -> int:
    """Approximate token count. Claude API gives exact counts."""
    return max(1, len(str(text)) // 4)


def count_tokens_list(items: list) -> int:
    return sum(count_tokens(json.dumps(item)) for item in items)


# =============================================================================
# 2. CACHE SIMULATOR
#    Mirrors the Claude API's cache behaviour:
#      - Tracks what content is cached (by hash of the prefix)
#      - Reports cache_creation vs cache_read
#      - Enforces minimum 1024 token requirement
#      - Expires entries after 1 hour
#
#    Transcript: "cache is stored for one hour only"
# =============================================================================

class CacheSimulator:
    """
    Simulates Claude API's prompt caching behaviour for Ollama.
    In Claude API this happens server-side invisibly.
    Here we make it visible so you can understand the mechanics.
    """
    CACHE_TTL      = 3600   # 1 hour in seconds
    MIN_TOKENS     = 1024   # minimum tokens to cache

    def __init__(self):
        self._store: dict[str, dict] = {}   # hash → {tokens, created_at}
        self.total_cache_writes = 0
        self.total_cache_reads  = 0
        self.total_tokens_saved = 0

    def _hash(self, content) -> str:
        return hashlib.md5(json.dumps(content, sort_keys=True).encode()).hexdigest()

    def check(self, content, token_count: int) -> str:
        """
        Check if content is cached. Returns:
          "write" → not cached, will write to cache
          "read"  → cached, will read from cache
          "skip"  → too small to cache (< 1024 tokens)
        """
        if token_count < self.MIN_TOKENS:
            return "skip"

        key   = self._hash(content)
        entry = self._store.get(key)

        if entry:
            # Check TTL
            if time.time() - entry["created_at"] < self.CACHE_TTL:
                return "read"
            else:
                del self._store[key]   # expired

        return "write"

    def write(self, content, token_count: int):
        key = self._hash(content)
        self._store[key] = {"tokens": token_count, "created_at": time.time()}
        self.total_cache_writes += token_count

    def read(self, content) -> int:
        key   = self._hash(content)
        entry = self._store.get(key, {})
        tokens = entry.get("tokens", 0)
        self.total_cache_reads  += tokens
        self.total_tokens_saved += tokens
        return tokens

    def stats(self) -> dict:
        return {
            "cache_writes": self.total_cache_writes,
            "cache_reads":  self.total_cache_reads,
            "tokens_saved": self.total_tokens_saved,
        }


# =============================================================================
# 3. TOOL SCHEMAS
#    Transcript: "add cache_control to the very last tool schema"
#    We store cache_control as metadata — stripped before sending to Ollama,
#    kept for Claude API.
# =============================================================================

calculator_schema = {
    "type": "function",
    "function": {
        "name":        "calculator",
        "description": "Evaluate a math expression.",
        "parameters":  {
            "type":       "object",
            "properties": {"expression": {"type": "string"}},
            "required":   ["expression"],
        },
    },
}

get_time_schema = {
    "type": "function",
    "function": {
        "name":        "get_current_time",
        "description": "Get the current time.",
        "parameters":  {"type": "object", "properties": {}, "required": []},
    },
}

set_reminder_schema = {
    "type": "function",
    "function": {
        "name":        "set_reminder",
        "description": "Set a reminder at a given date.",
        "parameters":  {
            "type":       "object",
            "properties": {
                "date_str": {"type": "string"},
                "content":  {"type": "string"},
            },
            "required": ["date_str", "content"],
        },
    },
    # ── Cache breakpoint on the LAST tool schema ──────────────────────────
    # Transcript: "add cache_control to the very last tool in the list"
    # Claude API uses this field natively.
    # We use it as a marker for our simulator.
    "cache_control": {"type": "ephemeral"},
}

ALL_TOOLS = [calculator_schema, get_time_schema, set_reminder_schema]

# Large system prompt — transcript used a 6K token prompt to demo caching
# We use a shorter one but the mechanics are identical
SYSTEM_PROMPT = """You are a helpful assistant with access to tools.
Use tools when appropriate. Always be concise and accurate.
When asked about math, use the calculator tool.
When asked about time, use the get_current_time tool.
When asked to set reminders, use the set_reminder tool.
For general questions, answer directly without tools.
""" * 20   # repeat to get closer to 1024 tokens for caching demo


# =============================================================================
# 4. CHAT FUNCTION WITH CACHING
#    Mirrors the transcript's updated chat() function that:
#      1. Clones the tools list (don't mutate the original)
#      2. Adds cache_control to the last tool
#      3. Adds cache_control to the system prompt
#      4. Reports cache usage in the response
#
#    For Ollama: strips cache_control before sending, simulates the behaviour
#    For Claude: passes cache_control through natively
# =============================================================================

_cache = CacheSimulator()

def chat(
    messages:      list,
    tools:         list = None,
    system:        str  = None,
    enable_cache:  bool = True,
    model:         str  = MODEL,
) -> dict:
    """
    Chat with optional prompt caching.

    Returns:
      {
        "content": str,
        "usage": {
            "input_tokens":               int,
            "output_tokens":              int,
            "cache_creation_input_tokens": int,   # tokens written to cache
            "cache_read_input_tokens":     int,   # tokens read from cache
        }
      }

    Mirrors transcript's chat() that always caches tools + system prompt.
    """
    if MODE == "claude":
        return _chat_claude(messages, tools, system, enable_cache)
    else:
        return _chat_ollama(messages, tools, system, enable_cache, model)


def _chat_ollama(messages, tools, system, enable_cache, model):
    """Ollama implementation with cache simulation."""

    usage = {
        "input_tokens":                0,
        "output_tokens":               0,
        "cache_creation_input_tokens": 0,
        "cache_read_input_tokens":     0,
    }

    # ── Simulate tool caching ─────────────────────────────────────────────
    # Transcript: "clone the tools list, add cache_control to the last item"
    ollama_tools = None
    if tools and enable_cache:
        # Clone so we don't mutate the original — transcript's good practice tip
        tools_clone = [t.copy() for t in tools]
        last_tool   = tools_clone[-1].copy()
        last_tool["cache_control"] = {"type": "ephemeral"}
        tools_clone[-1] = last_tool

        tool_tokens = count_tokens_list(tools)
        status      = _cache.check(tools, tool_tokens)

        if status == "write":
            _cache.write(tools, tool_tokens)
            usage["cache_creation_input_tokens"] += tool_tokens
            print(f"  [cache] tools → WRITE ({tool_tokens} tokens)")
        elif status == "read":
            saved = _cache.read(tools)
            usage["cache_read_input_tokens"] += saved
            print(f"  [cache] tools → READ  ({saved} tokens saved)")
        else:
            print(f"  [cache] tools → SKIP  (< {CacheSimulator.MIN_TOKENS} tokens)")

        # Strip cache_control before sending to Ollama
        ollama_tools = [
            {k: v for k, v in t.items() if k != "cache_control"}
            for t in tools_clone
        ]

    # ── Simulate system prompt caching ───────────────────────────────────
    # Transcript: "replace system string with a list containing a text block
    #              with cache_control"
    ollama_messages = list(messages)
    if system and enable_cache:
        sys_tokens = count_tokens(system)
        status     = _cache.check(system, sys_tokens)

        if status == "write":
            _cache.write(system, sys_tokens)
            usage["cache_creation_input_tokens"] += sys_tokens
            print(f"  [cache] system → WRITE ({sys_tokens} tokens)")
        elif status == "read":
            saved = _cache.read(system)
            usage["cache_read_input_tokens"] += saved
            print(f"  [cache] system → READ  ({saved} tokens saved)")
        else:
            print(f"  [cache] system → SKIP  (< {CacheSimulator.MIN_TOKENS} tokens)")

        # Inject system as first message for Ollama
        ollama_messages = [
            {"role": "system", "content": system}
        ] + ollama_messages

    # ── Count input tokens ────────────────────────────────────────────────
    usage["input_tokens"] = count_tokens_list(ollama_messages)

    # ── Send to Ollama ────────────────────────────────────────────────────
    payload = {"model": model, "messages": ollama_messages, "stream": False}
    if ollama_tools:
        payload["tools"] = ollama_tools

    res = requests.post(OLLAMA_URL, json=payload)
    res.raise_for_status()

    content = res.json()["message"]["content"]
    usage["output_tokens"] = count_tokens(content)

    return {"content": content, "usage": usage}


def _chat_claude(messages, tools, system, enable_cache):
    """
    Real Claude API implementation with actual caching.
    Switch MODE = "claude" and set ANTHROPIC_API_KEY to use this.

    This is the direct translation of the transcript's notebook code.
    """
    import anthropic
    client = anthropic.Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

    params = {"model": "claude-opus-4-5", "max_tokens": 1024, "messages": messages}

    # Cache tools — transcript: "clone list, add cache_control to last tool"
    if tools and enable_cache:
        tools_clone     = [t.copy() for t in tools]
        last            = tools_clone[-1].copy()
        last["cache_control"] = {"type": "ephemeral"}
        tools_clone[-1] = last
        params["tools"] = tools_clone

    # Cache system prompt — transcript: "wrap in list with cache_control"
    if system and enable_cache:
        params["system"] = [
            {"type": "text", "text": system, "cache_control": {"type": "ephemeral"}}
        ]

    params["betas"] = ["prompt-caching-2024-07-31"]

    response = client.beta.prompt_caching.messages.create(**params)
    usage    = response.usage

    return {
        "content": response.content[0].text,
        "usage": {
            "input_tokens":                usage.input_tokens,
            "output_tokens":               usage.output_tokens,
            "cache_creation_input_tokens": getattr(usage, "cache_creation_input_tokens", 0),
            "cache_read_input_tokens":     getattr(usage, "cache_read_input_tokens",     0),
        },
    }


# =============================================================================
# 5. USAGE REPORTER
#    Transcript: "take a look at the usage field — cache_creation vs cache_read"
# =============================================================================

def print_usage(usage: dict, label: str = ""):
    print(f"\n  {'─'*40}")
    if label:
        print(f"  {label}")
    print(f"  input_tokens:                {usage['input_tokens']}")
    print(f"  output_tokens:               {usage['output_tokens']}")
    print(f"  cache_creation_input_tokens: {usage['cache_creation_input_tokens']}")
    print(f"  cache_read_input_tokens:     {usage['cache_read_input_tokens']}")


# =============================================================================
# 6. DEMO
#    Mirrors the transcript's notebook walkthrough exactly:
#    1. No tools, no system → baseline
#    2. Add tools → cache WRITE
#    3. Same tools again → cache READ
#    4. Change user message → still cache READ (prefix unchanged)
#    5. Change a tool → cache MISS, back to WRITE
#    6. Add system prompt → partial read + new write
# =============================================================================

if __name__ == "__main__":

    print("=" * 55)
    print(f"MODE: {MODE}")
    print("=" * 55)

    # ── Step 1: No tools, no system — baseline
    # Transcript: "pass in nothing at all, see the baseline token count"
    print("\n[Step 1] Baseline — no tools, no system prompt")
    result = chat(
        messages = [{"role": "user", "content": "What is 1+1?"}],
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Baseline")

    # ── Step 2: Add tools → cache WRITE
    # Transcript: "add tools, see cache_creation_input_tokens"
    print("\n[Step 2] Add tools → expect cache WRITE")
    result = chat(
        messages = [{"role": "user", "content": "What is 1+1?"}],
        tools    = ALL_TOOLS,
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Tools added — first request")

    # ── Step 3: Same tools, same question → cache READ
    # Transcript: "make the same follow-up request, see cache_read_input_tokens"
    print("\n[Step 3] Same tools, same question → expect cache READ")
    result = chat(
        messages = [{"role": "user", "content": "What is 1+1?"}],
        tools    = ALL_TOOLS,
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Same tools — second request")

    # ── Step 4: Same tools, DIFFERENT question → still cache READ
    # Transcript: "change the user message — still reads from cache"
    print("\n[Step 4] Same tools, different question → still cache READ")
    result = chat(
        messages = [{"role": "user", "content": "What is 99 * 42?"}],
        tools    = ALL_TOOLS,
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Different question — cache still hits")

    # ── Step 5: Change a tool → cache MISS
    # Transcript: "change the description on the first tool → invalidate cache"
    print("\n[Step 5] Modified tool → expect cache MISS (back to WRITE)")
    modified_tools = [t.copy() for t in ALL_TOOLS]
    modified_tools[0] = modified_tools[0].copy()
    modified_tools[0]["function"] = modified_tools[0]["function"].copy()
    modified_tools[0]["function"]["description"] = "Evaluate a math expression (updated)"

    result = chat(
        messages = [{"role": "user", "content": "What is 1+1?"}],
        tools    = modified_tools,
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Modified tool — cache invalidated")

    # ── Step 6: Add system prompt → partial read + new write
    # Transcript: "add system prompt — expect cache_read for tools, cache_write for system"
    print("\n[Step 6] Original tools + system prompt → READ tools, WRITE system")
    result = chat(
        messages = [{"role": "user", "content": "What is 1+1?"}],
        tools    = ALL_TOOLS,
        system   = SYSTEM_PROMPT,
    )
    print(f"  Answer: {result['content'][:80]}")
    print_usage(result["usage"], "Tools + system prompt")

    # ── Final cache stats
    print("\n" + "=" * 55)
    print("CACHE SUMMARY")
    print("=" * 55)
    stats = _cache.stats()
    print(f"  Total tokens written to cache: {stats['cache_writes']}")
    print(f"  Total tokens read from cache:  {stats['cache_reads']}")
    print(f"  Total tokens saved:            {stats['tokens_saved']}")
    print(f"\n  In Claude API, cached tokens cost ~10% of normal input price.")
    print(f"  Cache writes cost ~125% — but pay off after the second request.")
````

# Code execution and the Files API

````python
"""
files_and_code_execution.py
---------------------------
Translates "Code Execution and the Files API" transcript to Ollama/Llama.

Claude API:
  Files API:
    - Upload a file once → get file_id
    - Reference file_id in future requests instead of re-sending raw bytes
    - Download files generated inside Docker container via file_id

  Code Execution:
    - Pass a predefined schema → Claude runs Python in isolated Docker
    - No network access inside container
    - Files go in/out via Files API
    - Claude can run code multiple times, each result fed back to Claude

Ollama translation:
  Files API:
    - We manage file storage ourselves: {file_id: local_path}
    - UUID as file_id, local path as value
    - upload/list/delete/download functions mirror the transcript's helpers

  Code Execution:
    - LLM writes Python code
    - We execute it with subprocess (our "Docker container")
    - Capture stdout/stderr/return_code
    - Inject output back into conversation
    - Loop until LLM stops writing code
    - Plots saved locally, "downloaded" via file_id
"""

import json
import os
import re
import subprocess
import tempfile
import uuid
import requests
import pandas as pd
import numpy as np

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL      = "qwen2.5:7b"


# =============================================================================
# 1. FILES API SIMULATION
#    Mirrors the transcript's helper functions:
#      upload(), list_files(), delete_file(), download_file(), get_file_info()
#
#    Claude API: HTTP requests to api.anthropic.com/v1/files
#    Ollama:     in-memory dict mapping file_id → local path
# =============================================================================

class FilesAPI:
    """
    Simulates Anthropic's Files API for local file management.
    Transcript: "upload a file, get back a File Metadata Object with a file_id"
    """

    def __init__(self):
        self._store: dict[str, dict] = {}   # file_id → {path, name, size}

    def upload(self, file_path: str) -> dict:
        """
        Upload a file and return a File Metadata Object.
        Transcript: "make an initial request to upload a document, get back a file_id"
        """
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"File not found: {file_path}")

        file_id  = str(uuid.uuid4())
        name     = os.path.basename(file_path)
        size     = os.path.getsize(file_path)

        self._store[file_id] = {
            "file_id":  file_id,
            "name":     name,
            "path":     file_path,
            "size":     size,
        }

        print(f"  [Files API] Uploaded: {name} → file_id: {file_id[:8]}...")
        return self._store[file_id]

    def list_files(self) -> list[dict]:
        """List all uploaded files. Transcript: "list_files function"."""
        return list(self._store.values())

    def get_file_info(self, file_id: str) -> dict:
        """Get metadata for a specific file."""
        if file_id not in self._store:
            raise ValueError(f"File not found: {file_id}")
        return self._store[file_id]

    def delete_file(self, file_id: str) -> bool:
        """Delete a file from the store. Transcript: "we can delete a file"."""
        if file_id in self._store:
            name = self._store[file_id]["name"]
            del self._store[file_id]
            print(f"  [Files API] Deleted: {name}")
            return True
        return False

    def download_file(self, file_id: str, save_path: str = None) -> str:
        """
        Download a file by file_id.
        Transcript: "use download_file to get the generated plot out of the container"
        """
        info = self.get_file_info(file_id)
        src  = info["path"]

        if save_path is None:
            save_path = f"/tmp/downloaded_{info['name']}"

        import shutil
        shutil.copy2(src, save_path)
        print(f"  [Files API] Downloaded: {info['name']} → {save_path}")
        return save_path

    def get_path(self, file_id: str) -> str:
        """Get local path for a file_id — used internally by code executor."""
        return self.get_file_info(file_id)["path"]


# =============================================================================
# 2. CODE EXECUTION TOOL
#    Mirrors Claude's server-side code execution in a Docker container.
#
#    Claude API:
#      - Pass predefined schema → Claude writes and runs Python in Docker
#      - stdout/stderr/return_code returned as tool result
#      - No network access inside container
#      - Files injected via container_upload block
#
#    Ollama:
#      - LLM writes Python code
#      - We execute with subprocess (our "container")
#      - Inject available files as local paths the code can read
#      - Capture all output
# =============================================================================

CODE_EXECUTION_SCHEMA = {
    "type": "function",
    "function": {
        "name":        "execute_python",
        "description": (
            "Execute Python code to analyse data, generate plots, or perform calculations. "
            "Available libraries: pandas, numpy, matplotlib, scipy. "
            "Use print() to output results. "
            "Files are available at the paths provided in the system context. "
            "Save plots using plt.savefig(filename)."
        ),
        "parameters": {
            "type":       "object",
            "properties": {
                "code": {
                    "type":        "string",
                    "description": "The Python code to execute.",
                },
                "reason": {
                    "type":        "string",
                    "description": "Brief explanation of what this code does.",
                },
            },
            "required": ["code"],
        },
    },
}


class CodeExecutor:
    """
    Executes Python code in an isolated subprocess.
    Mirrors Claude's Docker container execution environment.

    Transcript: "Claude can run code inside the container multiple times.
                 Whatever Claude prints will be sent back to Claude."
    """

    def __init__(self, files_api: FilesAPI, work_dir: str = "/tmp/code_exec"):
        self.files_api = files_api
        self.work_dir  = work_dir
        os.makedirs(work_dir, exist_ok=True)
        self._generated_files: list[str] = []   # files created during execution

    def execute(self, code: str, available_file_ids: list[str] = None) -> dict:
        """
        Execute Python code and return structured result.
        Mirrors transcript's code_execution_tool_result block:
          stdout, stderr, return_code
        """
        # Inject file paths into code context
        file_context = self._build_file_context(available_file_ids or [])

        # Prepend file context and work_dir setup to the code
        full_code = f"""
import os
import warnings
warnings.filterwarnings('ignore')
os.chdir("{self.work_dir}")

# Available files (injected by executor — mirrors container_upload block)
{file_context}

# User code
{code}
"""
        # Write to temp file and execute
        code_file = os.path.join(self.work_dir, "exec_code.py")
        with open(code_file, "w") as f:
            f.write(full_code)

        result = subprocess.run(
            ["python3", code_file],
            capture_output = True,
            text           = True,
            timeout        = 30,
            cwd            = self.work_dir,
        )

        # Collect any files generated during execution
        generated = self._collect_generated_files()

        return {
            "stdout":      result.stdout,
            "stderr":      result.stderr,
            "return_code": result.returncode,
            "generated":   generated,   # files created (plots etc.)
        }

    def _build_file_context(self, file_ids: list[str]) -> str:
        """
        Build Python variable assignments for available files.
        Mirrors the transcript's container_upload block — injecting files
        into the container so the code can access them.
        """
        lines = []
        for fid in file_ids:
            try:
                info = self.files_api.get_file_info(fid)
                var  = re.sub(r"[^a-zA-Z0-9_]", "_", info["name"].split(".")[0])
                lines.append(f'{var}_path = "{info["path"]}"')
                lines.append(f'# File: {info["name"]} available as {var}_path')
            except ValueError:
                pass
        return "\n".join(lines) if lines else "# No files injected"

    def _collect_generated_files(self) -> list[str]:
        """Find files created during execution — mirrors plot download via file_id."""
        generated = []
        for fname in os.listdir(self.work_dir):
            full = os.path.join(self.work_dir, fname)
            if fname != "exec_code.py" and os.path.isfile(full):
                if full not in self._generated_files:
                    self._generated_files.append(full)
                    generated.append(full)
        return generated

    def register_output_files(self, generated: list[str]) -> list[str]:
        """
        Register generated files with Files API and return their file_ids.
        Transcript: "the plot is stored inside the Docker container —
                     use the file_id to download it"
        """
        file_ids = []
        for path in generated:
            meta = self.files_api.upload(path)
            file_ids.append(meta["file_id"])
        return file_ids


# =============================================================================
# 3. CHAT WITH CODE EXECUTION
#    Multi-turn loop:
#      LLM writes code → we execute → inject result → LLM writes more or answers
#
#    Transcript: "Claude can run code multiple times inside the container.
#                 Whatever Claude prints will be sent back to Claude."
# =============================================================================

def build_system_prompt(file_ids: list[str], files_api: FilesAPI) -> str:
    """Build system prompt with explicit variable names so LLM uses them correctly."""
    if not file_ids:
        return "You are a data analyst assistant. Use execute_python to analyse data."

    file_lines = []
    for fid in file_ids:
        info    = files_api.get_file_info(fid)
        varname = re.sub(r"[^a-zA-Z0-9_]", "_", info["name"].split(".")[0])
        file_lines.append(
            f"  - {info['name']} is available as the variable: {varname}_path\n"
            f"    To load it: df = pd.read_csv({varname}_path)"
        )

    file_vars = "\n".join(
        f"{re.sub(r'[^a-zA-Z0-9_]', '_', info['name'].split('.')[0])}_path = available"
        for info in [files_api.get_file_info(fid) for fid in file_ids]
    )
    return (
        "You are a data analyst. Use execute_python for ALL analysis.\n"
        "Always use print() for output. If code fails, fix and retry.\n"
        f"File variables available: {file_vars}"
    )


def run_code_execution(
    user_prompt:      str,
    file_ids:         list[str],
    files_api:        FilesAPI,
    executor:         CodeExecutor,
    model:            str = MODEL,
) -> dict:
    """
    Full code execution pipeline.
    Mirrors transcript's complete flow:
      upload → container_upload block → LLM analyses → download output
    """
    print(f"\n{'='*55}")
    print(f"USER: {user_prompt}")
    print(f"{'='*55}")

    system   = build_system_prompt(file_ids, files_api)
    messages = [
        {"role": "system", "content": system},
        {"role": "user",   "content": user_prompt},
    ]

    all_generated_file_ids = []
    execution_log          = []
    max_iterations         = 12

    for iteration in range(max_iterations):

        payload = {
            "model":    model,
            "messages": messages,
            "stream":   False,
            "tools":    [CODE_EXECUTION_SCHEMA],
        }
        res = requests.post(OLLAMA_URL, json=payload)
        res.raise_for_status()
        response = res.json()["message"]
        messages.append(response)

        tool_calls = response.get("tool_calls")

        # ── No tool call → LLM is done ────────────────────────────────────
        if not tool_calls:
            final_answer = response.get("content", "")
            print(f"\nFINAL ANSWER:\n{final_answer}")

            return {
                "answer":        final_answer,
                "execution_log": execution_log,
                "output_files":  all_generated_file_ids,
            }

        # ── Execute code ──────────────────────────────────────────────────
        for call in tool_calls:
            fn     = call["function"]
            args   = fn.get("arguments", {})
            if isinstance(args, str):
                try:
                    args = json.loads(args)
                except json.JSONDecodeError:
                    args = {}

            code   = args.get("code", "")
            reason = args.get("reason", "")

            print(f"\n[CODE EXECUTION {iteration+1}] {reason}")
            print(f"{'─'*40}")
            print(code[:400] + ("..." if len(code) > 400 else ""))

            # Execute — mirrors running code in Docker container
            result = executor.execute(code, file_ids)

            print(f"\n[STDOUT]\n{result['stdout'][:500]}" if result["stdout"] else "")
            if result["stderr"]:
                print(f"[STDERR]\n{result['stderr'][:200]}")
            print(f"[return_code] {result['return_code']}")

            # Register any generated files (plots etc.)
            if result["generated"]:
                new_ids = executor.register_output_files(result["generated"])
                all_generated_file_ids.extend(new_ids)
                print(f"[Generated files] {[files_api.get_file_info(fid)['name'] for fid in new_ids]}")

            # Log execution
            execution_log.append({
                "iteration": iteration + 1,
                "code":      code,
                "stdout":    result["stdout"],
                "stderr":    result["stderr"],
                "return_code": result["return_code"],
            })

            # Inject result back into conversation
            # Mirrors transcript's code_execution_tool_result block
            tool_result = (
                f"EXECUTION RESULT (return_code={result['return_code']}):\n"
                f"stdout:\n{result['stdout']}\n"
            )
            if result["stderr"]:
                tool_result += f"stderr:\n{result['stderr']}\n"
            if result["generated"]:
                tool_result += f"Generated files: {[os.path.basename(p) for p in result['generated']]}\n"

            messages.append({"role": "tool", "content": tool_result})

    print("⚠ Max iterations reached.")
    return {"answer": "", "execution_log": execution_log, "output_files": all_generated_file_ids}


# =============================================================================
# 4. SAMPLE DATA GENERATOR
#    Mirrors transcript's streaming.csv:
#    user data with churn column for analysis demo
# =============================================================================

def create_sample_csv(path: str) -> str:
    """
    Create a sample streaming service CSV matching the transcript's demo.
    Columns: user_id, subscription_tier, hours_viewed, top_genre,
             monthly_cost, support_tickets, churned
    """
    np.random.seed(42)
    n = 200

    tiers   = np.random.choice(["basic", "standard", "premium"], n, p=[0.4, 0.4, 0.2])
    costs   = {"basic": 9.99, "standard": 14.99, "premium": 19.99}
    hours   = np.random.exponential(40, n).clip(1, 200)
    genres  = np.random.choice(["Drama", "Comedy", "Action", "Documentary", "Thriller"], n)
    tickets = np.random.poisson(1, n)

    # Churn probability — higher for low hours and high cost
    churn_prob = 0.3 - 0.002 * hours + 0.01 * tickets
    churn_prob = np.clip(churn_prob, 0.05, 0.85)
    churned    = np.random.binomial(1, churn_prob)

    df = pd.DataFrame({
        "user_id":          range(1, n + 1),
        "subscription_tier": tiers,
        "hours_viewed":     hours.round(1),
        "top_genre":        genres,
        "monthly_cost":     [costs[t] for t in tiers],
        "support_tickets":  tickets,
        "churned":          churned,
    })

    df.to_csv(path, index=False)
    print(f"  Created sample CSV: {path} ({n} rows)")
    return path


# =============================================================================
# 5. DEMO
# =============================================================================

if __name__ == "__main__":

    files_api = FilesAPI()
    executor  = CodeExecutor(files_api)
    # Clean work dir from any previous runs
    import shutil
    if os.path.exists("/tmp/code_exec"):
        shutil.rmtree("/tmp/code_exec")
    os.makedirs("/tmp/code_exec", exist_ok=True)

    # ── Step 1: Upload CSV — mirrors transcript's file upload
    print("=" * 55)
    print("STEP 1 — Upload file via Files API")
    print("=" * 55)

    csv_path = "/tmp/streaming.csv"
    create_sample_csv(csv_path)

    meta = files_api.upload(csv_path)
    file_id = meta["file_id"]
    print(f"\nFile Metadata Object:")
    print(json.dumps({k: v for k, v in meta.items() if k != "path"}, indent=2))

    # ── Step 2: List files
    print("\n" + "=" * 55)
    print("STEP 2 — List uploaded files")
    print("=" * 55)
    for f in files_api.list_files():
        print(f"  {f['file_id'][:8]}... | {f['name']} | {f['size']} bytes")

    # ── Step 3: Code execution — analysis
    # Transcript: "ask Claude to analyse why customers are cancelling"
    print("\n" + "=" * 55)
    print("STEP 3 — Code execution: churn analysis")
    print("=" * 55)

    result = run_code_execution(
        user_prompt = (
            "Load streaming.csv using pd.read_csv(streaming_path). "
            "Columns: user_id, subscription_tier, hours_viewed, monthly_cost, support_tickets, churned. "
            "Do these in order using execute_python:\n"
            "1. print(df.head()) and print(df.churned.value_counts())\n"
            "2. print(df.groupby('subscription_tier')['churned'].mean())\n"
            "3. print(df[['churned','hours_viewed','support_tickets','monthly_cost']].corr())\n"
            "4. Make and save bar chart of churn by tier as churn_plot.png\n"
            "5. Print your findings summary"
        ),
        file_ids  = [file_id],
        files_api = files_api,
        executor  = executor,
    )

    # ── Step 4: Download generated plot
    # Transcript: "use the file_id to download the generated plot"
    print("\n" + "=" * 55)
    print("STEP 4 — Download generated files")
    print("=" * 55)

    if result["output_files"]:
        for fid in result["output_files"]:
            info      = files_api.get_file_info(fid)
            save_path = f"/tmp/downloaded_{info['name']}"
            files_api.download_file(fid, save_path)
            print(f"  Plot saved to: {save_path}")
    else:
        print("  No output files generated.")

    # ── Step 5: Execution log
    # Transcript: "Claude ran some code several times in a row"
    print("\n" + "=" * 55)
    print("STEP 5 — Execution log")
    print("=" * 55)
    print(f"Total code executions: {len(result['execution_log'])}")
    for log in result["execution_log"]:
        print(f"\n  Iteration {log['iteration']} — return_code: {log['return_code']}")
        print(f"  Output preview: {log['stdout'][:150].strip()}...")
````
